```python
# ==============================================================================
# Google Colab Complete Verification Suite for HYPOTHESIS 2:
# Spinodal Instability in Battleground States, Inter-State Spatial Coupling
# (W_inter), and Industrial Contagion (Petro/ICE vs. NATO/Asia Supply Chains)
# Comparing 2018-2020 Split Congress vs. 2026-2028 Super-State & Leak Regimes
# - Strictly 6-Batch Vectorized Monte Carlo (1,200,000 to 10,000,000 Trials/Regime)
# - Automatic CSV, PNG Figures, LaTeX Table Snippet, and ZIP Browser Download
# ==============================================================================
import os
import time
import zipfile
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.special import erf

plt.rcParams['figure.dpi'] = 150
plt.rcParams['savefig.dpi'] = 300
plt.rcParams['font.sans-serif'] = 'DejaVu Sans'
plt.rcParams['axes.edgecolor'] = '#222222'
plt.rcParams['axes.linewidth'] = 0.85

OUTPUT_DIR = "Hypothesis2_Spinodal_Spatial_Contagion_Suite"
os.makedirs(OUTPUT_DIR, exist_ok=True)

# ==============================================================================
# SCALE CONFIGURATION:
# Default: 1,200,000 trials per regime (6 batches x 200,000 trials)
# For 10,000,000 trials per regime, set N_TRIALS_PER_REGIME = 9_999_996 (6 x 1,666,666)
# ==============================================================================
N_TRIALS_PER_REGIME = 1_200_000   # <-- Change to 9_999_996 for 10M trials per regime
N_BATCHES = 6                     # Strictly 6 independent batches / Markov chains
SEED_BASE = 20262200

print("=" * 82)
print(" HYPOTHESIS 2 SPINODAL INSTABILITY & SPATIAL CONTAGION MONTE CARLO SUITE")
print(f" Trials per Regime: {N_TRIALS_PER_REGIME:,} (6 Batches x {N_TRIALS_PER_REGIME // N_BATCHES:,})")
print("=" * 82)

# ==============================================================================
# STEP 1: 50-State Structural Parameters & Spatial Coupling Matrix W_inter
# ==============================================================================
def build_us50_spatial_network():
    raw_states = [
        ("PA", "Battleground", 19, 48.6, -0.015, 0.58, 0.41, 0.31, 0.28, +0.45, 0),
        ("MI", "Battleground", 15, 47.6, -0.022, 0.58, 0.42, 0.30, 0.28, +0.68, 1),
        ("WI", "Battleground", 10, 48.5, -0.020, 0.58, 0.42, 0.30, 0.28, +0.35, 0),
        ("GA", "Battleground", 16, 49.3, -0.005, 0.56, 0.40, 0.32, 0.28, +0.20, 1),
        ("NC", "Battleground", 16, 50.6, +0.006, 0.56, 0.41, 0.31, 0.28, +0.25, 1),
        ("AZ", "Battleground", 11, 49.2, -0.008, 0.57, 0.39, 0.31, 0.30, +0.10, 0),
        ("NV", "Battleground",  6, 48.1, -0.014, 0.59, 0.40, 0.30, 0.30, +0.15, 0),
        ("MN", "Lean", 10, 41.0, -0.085, 0.55, 0.36, 0.33, 0.31, +0.10, 1),
        ("NH", "Lean",  4, 42.9, -0.070, 0.56, 0.35, 0.33, 0.32, -0.20, 1),
        ("VA", "Lean", 13, 39.0, -0.115, 0.54, 0.34, 0.36, 0.30, -0.50, 1),
        ("ME", "Lean",  4, 41.3, -0.078, 0.57, 0.36, 0.32, 0.32, -0.15, 1),
        ("NM", "Lean",  5, 40.0, -0.095, 0.55, 0.36, 0.33, 0.31, +0.55, 1),
        ("OH", "Lean", 17, 61.7, +0.115, 0.52, 0.46, 0.28, 0.26, +0.62, 1),
        ("IA", "Lean",  6, 62.5, +0.120, 0.52, 0.46, 0.28, 0.26, +0.50, 1),
        ("FL", "Lean", 30, 59.8, +0.090, 0.51, 0.44, 0.29, 0.27, +0.30, 1),
        ("TX", "Lean", 40, 60.7, +0.098, 0.51, 0.45, 0.28, 0.27, +0.92, 1),
        ("AL", "Solid Red",  9, 81.8, +0.340, 0.40, 0.50, 0.26, 0.24, +0.65, 1),
        ("AK", "Solid Red",  3, 65.7, +0.155, 0.50, 0.46, 0.27, 0.27, +0.95, 1),
        ("AR", "Solid Red",  6, 83.8, +0.370, 0.39, 0.51, 0.25, 0.24, +0.60, 1),
        ("ID", "Solid Red",  4, 84.6, +0.395, 0.38, 0.51, 0.25, 0.24, +0.50, 1),
        ("IN", "Solid Red", 11, 72.0, +0.225, 0.45, 0.48, 0.27, 0.25, +0.68, 0),
        ("KS", "Solid Red",  6, 69.7, +0.200, 0.46, 0.47, 0.28, 0.25, +0.55, 1),
        ("KY", "Solid Red",  8, 81.8, +0.345, 0.40, 0.50, 0.26, 0.24, +0.72, 1),
        ("LA", "Solid Red",  8, 73.8, +0.240, 0.44, 0.48, 0.27, 0.25, +0.90, 1),
        ("MS", "Solid Red",  6, 73.0, +0.232, 0.44, 0.48, 0.27, 0.25, +0.65, 1),
        ("MO", "Solid Red", 10, 70.2, +0.205, 0.46, 0.47, 0.28, 0.25, +0.55, 0),
        ("MT", "Solid Red",  4, 71.7, +0.225, 0.45, 0.48, 0.27, 0.25, +0.70, 1),
        ("NE", "Solid Red",  5, 73.8, +0.245, 0.44, 0.48, 0.27, 0.25, +0.55, 1),
        ("ND", "Solid Red",  3, 84.6, +0.402, 0.38, 0.52, 0.24, 0.24, +0.95, 0),
        ("OK", "Solid Red",  7, 84.6, +0.415, 0.37, 0.52, 0.24, 0.24, +0.92, 1),
        ("SC", "Solid Red",  9, 67.7, +0.180, 0.47, 0.47, 0.28, 0.25, +0.55, 1),
        ("SD", "Solid Red",  3, 81.8, +0.340, 0.40, 0.50, 0.26, 0.24, +0.60, 1),
        ("TN", "Solid Red", 11, 80.2, +0.335, 0.40, 0.50, 0.26, 0.24, +0.65, 1),
        ("UT", "Solid Red",  6, 75.8, +0.280, 0.43, 0.49, 0.26, 0.25, +0.50, 0),
        ("WV", "Solid Red",  4, 84.6, +0.460, 0.36, 0.53, 0.24, 0.23, +0.95, 1),
        ("WY", "Solid Red",  3, 84.6, +0.490, 0.35, 0.54, 0.23, 0.23, +0.95, 1),
        ("CA", "Solid Blue", 54, 16.0, -0.390, 0.42, 0.28, 0.38, 0.34, -0.88, 0),
        ("CO", "Solid Blue", 10, 34.4, -0.175, 0.52, 0.32, 0.35, 0.33, -0.25, 1),
        ("CT", "Solid Blue",  7, 26.0, -0.295, 0.46, 0.30, 0.37, 0.33, -0.68, 0),
        ("DE", "Solid Blue",  3, 28.0, -0.265, 0.47, 0.31, 0.36, 0.33, -0.52, 1),
        ("HI", "Solid Blue",  4, 16.0, -0.382, 0.42, 0.28, 0.38, 0.34, -0.75, 0),
        ("IL", "Solid Blue", 19, 28.0, -0.258, 0.47, 0.31, 0.36, 0.33, -0.45, 1),
        ("MD", "Solid Blue", 10, 16.0, -0.400, 0.41, 0.28, 0.39, 0.33, -0.78, 0),
        ("MA", "Solid Blue", 11, 15.0, -0.435, 0.40, 0.27, 0.39, 0.34, -0.88, 1),
        ("NJ", "Solid Blue", 14, 30.0, -0.245, 0.48, 0.31, 0.36, 0.33, -0.58, 1),
        ("NY", "Solid Blue", 28, 21.7, -0.345, 0.44, 0.29, 0.38, 0.33, -0.82, 0),
        ("OR", "Solid Blue",  8, 30.2, -0.225, 0.49, 0.31, 0.35, 0.34, -0.65, 1),
        ("RI", "Solid Blue",  4, 21.9, -0.335, 0.44, 0.29, 0.38, 0.33, -0.72, 1),
        ("VT", "Solid Blue",  3, 15.0, -0.450, 0.40, 0.27, 0.39, 0.34, -0.88, 0),
        ("WA", "Solid Blue", 12, 26.1, -0.282, 0.46, 0.30, 0.36, 0.34, -0.78, 0),
    ]
    df = pd.DataFrame(raw_states, columns=[
        "State", "Tier", "EV", "R_emp_pct", "b_k", "s_swing",
        "w_work", "w_est", "w_truth", "E_petro", "Senate_Contest_2026"
    ])
    # Vulnerability Index V_spinodal = s_swing / (|b_k| + 0.05) measuring proximity to spinodal instability
    df["Spinodal_Vulnerability_Idx"] = df["s_swing"] / (np.abs(df["b_k"]) + 0.05)
    df.to_csv(f"{OUTPUT_DIR}/01_US50_Spinodal_Vulnerability_and_Petro_Params.csv", index=False)

    b_vec = df["b_k"].values
    diff_mat = np.abs(b_vec[:, None] - b_vec[None, :])
    W_raw = np.exp(-diff_mat / 0.15)
    np.fill_diagonal(W_raw, 0.0)
    W_inter = W_raw / W_raw.sum(axis=1, keepdims=True)
    pd.DataFrame(W_inter, index=df["State"], columns=df["State"]).to_csv(
        f"{OUTPUT_DIR}/02_W_inter_50x50_Spatial_Coupling_Matrix.csv"
    )
    return df, W_inter

# ==============================================================================
# STEP 2: 6-Batch Vectorized Monte Carlo Engine for Hypothesis 2
# ==============================================================================
def run_hypothesis2_6batch_mc(regime_id, cfg, df_states, W_inter, n_trials=N_TRIALS_PER_REGIME, n_batches=N_BATCHES):
    t0 = time.time()
    K = len(df_states)
    batch_size = n_trials // n_batches
    total_N = batch_size * n_batches

    b_k = df_states["b_k"].values[None, :]
    s_swing = df_states["s_swing"].values[None, :]
    w_work = df_states["w_work"].values[None, :]
    w_est = df_states["w_est"].values[None, :]
    w_truth = df_states["w_truth"].values[None, :]
    E_petro = df_states["E_petro"].values[None, :]
    ev_vec = df_states["EV"].values
    senate_mask = df_states["Senate_Contest_2026"].values == 1

    beta = 2.45
    J_intra = cfg["J_intra"]
    J_inter = cfg["J_inter"]
    lam_damp = cfg["lam_damp"]
    q_zealot = cfg["q_zealot"]

    sum_X0, sum_X0_sq, sum_X0_m = [np.zeros(K, dtype=np.float64) for _ in range(3)]
    M1, M2, M3, M4 = [np.zeros(K, dtype=np.float64) for _ in range(4)]
    sum_q, sum_qw, sum_qe, sum_qt = [np.zeros(K, dtype=np.float64) for _ in range(4)]
    sum_Dw, sum_De, sum_Dt = [np.zeros(K, dtype=np.float64) for _ in range(3)]
    sum_soc_m = np.zeros(K, dtype=np.float64)
    sum_R, sum_R_sq, sum_win = [np.zeros(K, dtype=np.float64) for _ in range(3)]

    batch_means_m = np.zeros((n_batches, K), dtype=np.float64)
    batch_vars_m = np.zeros((n_batches, K), dtype=np.float64)
    ev_samples = []

    for b_idx in range(n_batches):
        rng = np.random.default_rng(SEED_BASE + regime_id * 1000 + b_idx)

        # Continuous uniform initial prior X0 ~ U(0, 0.2)
        X0 = rng.uniform(0.0, 0.2, size=(batch_size, K))
        m_curr = X0.copy()
        q_curr = np.full((batch_size, K), 0.85, dtype=np.float64)

        sum_X0 += X0.sum(axis=0)
        sum_X0_sq += (X0 ** 2).sum(axis=0)

        mode = cfg["mode"]
        if mode == "2018_split":
            xi_prot = rng.gamma(3.0, 0.12, size=(batch_size, 1))
            xi_econ = rng.normal(+0.18, 0.04, size=(batch_size, 1))
            eta_h = rng.normal(0.0, 0.025, size=(batch_size, K))
            eta_D = rng.normal(0.0, 0.025, size=(batch_size, K))
            delta_h = -0.0756

            for _ in range(12):
                soc_m = J_intra * m_curr + J_inter * (m_curr @ W_inter.T)
                h_w = b_k + 0.32 + 0.48 * xi_econ - 0.04 * xi_prot + eta_h
                D_w = 0.55 + 0.22 * xi_econ + eta_D
                h_e = b_k - 0.20 + 0.40 * delta_h - 0.15 * xi_prot + 0.12 * xi_econ + eta_h
                D_e = 0.31 + 0.10 * xi_prot + eta_D
                h_t = b_k - 0.16 + 0.50 * delta_h - 0.20 * xi_prot + eta_h
                D_t = 0.38 + 0.12 * xi_prot + eta_D

                aw, ae, at = beta * (soc_m + h_w), beta * (soc_m + h_e), beta * (soc_m + h_t)
                dw, de, dt = np.exp(-beta * D_w) + 2 * np.cosh(aw), np.exp(-beta * D_e) + 2 * np.cosh(ae), np.exp(-beta * D_t) + 2 * np.cosh(at)
                mw, qw = 2 * np.sinh(aw) / dw, 2 * np.cosh(aw) / dw
                me, qe = 2 * np.sinh(ae) / de, 2 * np.cosh(ae) / de
                mt, qt = 2 * np.sinh(at) / dt, 2 * np.cosh(at) / dt

                m_curr = (1.0 - lam_damp) * m_curr + lam_damp * ((1.0 - s_swing) * b_k + s_swing * (w_work * mw + w_est * me + w_truth * mt))
                q_curr = (1.0 - lam_damp) * q_curr + lam_damp * ((1.0 - s_swing) * q_zealot + s_swing * (w_work * qw + w_est * qe + w_truth * qt))

        elif mode == "2026_flip":
            xi_prot = rng.gamma(4.5, 0.22, size=(batch_size, 1))
            xi_econ = rng.normal(-0.12, 0.045, size=(batch_size, 1))
            eta_h = rng.normal(0.0, 0.028, size=(batch_size, K))
            eta_D = rng.normal(0.0, 0.030, size=(batch_size, K))
            delta_h = -0.1656

            for _ in range(12):
                soc_m = J_intra * m_curr + J_inter * (m_curr @ W_inter.T)
                h_w = b_k + 0.22 + 0.45 * xi_econ - 0.05 * xi_prot + eta_h
                D_w = 0.15 - 0.38 * xi_prot + eta_D
                h_e = b_k - 0.52 + 0.55 * delta_h - 0.22 * xi_prot + eta_h
                D_e = 0.78 + 0.35 * xi_prot + eta_D
                h_t = b_k - 0.38 + 0.85 * delta_h - 0.28 * xi_prot + eta_h
                D_t = 0.52 + 0.20 * xi_prot + eta_D

                aw, ae, at = beta * (soc_m + h_w), beta * (soc_m + h_e), beta * (soc_m + h_t)
                dw, de, dt = np.exp(-beta * D_w) + 2 * np.cosh(aw), np.exp(-beta * D_e) + 2 * np.cosh(ae), np.exp(-beta * D_t) + 2 * np.cosh(at)
                mw, qw = 2 * np.sinh(aw) / dw, 2 * np.cosh(aw) / dw
                me, qe = 2 * np.sinh(ae) / de, 2 * np.cosh(ae) / de
                mt, qt = 2 * np.sinh(at) / dt, 2 * np.cosh(at) / dt

                m_curr = (1.0 - lam_damp) * m_curr + lam_damp * ((1.0 - s_swing) * b_k + s_swing * (w_work * mw + w_est * me + w_truth * mt))
                q_curr = (1.0 - lam_damp) * q_curr + lam_damp * ((1.0 - s_swing) * q_zealot + s_swing * (w_work * qw + w_est * qe + w_truth * qt))

        else:
            xi_war   = rng.gamma(2.5, 0.22, size=(batch_size, 1))
            xi_ep    = rng.gamma(3.0, 0.28, size=(batch_size, 1))
            xi_tri   = rng.gamma(2.8, 0.24, size=(batch_size, 1))
            xi_panic = rng.gamma(3.5, 0.25, size=(batch_size, 1))

            eta_h = rng.normal(0.0, 0.028, size=(batch_size, K))
            eta_D = rng.normal(0.0, 0.030, size=(batch_size, K))

            M_w = np.zeros((batch_size, K), dtype=np.float64)
            M_e = np.zeros((batch_size, K), dtype=np.float64)
            M_t = np.zeros((batch_size, K), dtype=np.float64)

            alpha_mem = cfg["alpha_mem"]
            g_diss = cfg["gamma_diss"]
            ep_w_pen = cfg["ep_work_pen"]
            ep_e_pen = cfg["ep_est_pen"]
            petro_w = cfg["petro_weight"]

            psi_diss = g_diss * np.maximum(0.0, xi_tri - 0.45) ** 2

            for ww, we, wt, wp in cfg["stages"]:
                for _ in range(4):
                    soc_m = J_intra * m_curr + J_inter * (m_curr @ W_inter.T)
                    delta_h_petro = petro_w * (0.28 * E_petro * wp * xi_panic - 0.22 * (1.0 - E_petro) * ww * xi_war)

                    h_w = (b_k + 0.50 + 0.14 * ww * xi_war + 0.22 * wt * xi_tri + 0.18 * wp * xi_panic
                           + delta_h_petro - 0.55 * psi_diss - 0.42 * ep_w_pen * we * xi_ep + eta_h)
                    D_w = (0.65 + (0.26 + 0.15 * petro_w * E_petro) * wp * xi_panic
                           - 0.85 * psi_diss - ep_w_pen * we * xi_ep - 0.35 * M_w + eta_D)

                    h_e = b_k - 0.35 + 0.10 * ww * xi_war + 0.12 * wt * xi_tri + 0.08 * wp * xi_panic + 0.45 * delta_h_petro + eta_h
                    D_e = 0.44 - 0.40 * ww * xi_war - ep_e_pen * we * xi_ep - 0.46 * wt * xi_tri - 0.50 * wp * xi_panic - 0.38 * M_e + eta_D

                    h_t = b_k - 0.18 + 0.06 * ww * xi_war - 0.05 * we * xi_ep + eta_h
                    D_t = 0.32 - 0.28 * ww * xi_war - 1.15 * we * xi_ep - 0.25 * wt * xi_tri - 0.38 * wp * xi_panic - 0.42 * M_t + eta_D

                    aw, ae, at = beta * (soc_m + h_w), beta * (soc_m + h_e), beta * (soc_m + h_t)
                    dw, de, dt = np.exp(-beta * D_w) + 2 * np.cosh(aw), np.exp(-beta * D_e) + 2 * np.cosh(ae), np.exp(-beta * D_t) + 2 * np.cosh(at)
                    mw, qw = 2 * np.sinh(aw) / dw, 2 * np.cosh(aw) / dw
                    me, qe = 2 * np.sinh(ae) / de, 2 * np.cosh(ae) / de
                    mt, qt = 2 * np.sinh(at) / dt, 2 * np.cosh(at) / dt

                    M_w = alpha_mem * M_w + (1.0 - alpha_mem) * (1.0 - qw)
                    M_e = alpha_mem * M_e + (1.0 - alpha_mem) * (1.0 - qe)
                    M_t = alpha_mem * M_t + (1.0 - alpha_mem) * (1.0 - qt)

                    m_curr = (1.0 - lam_damp) * m_curr + lam_damp * ((1.0 - s_swing) * b_k + s_swing * (w_work * mw + w_est * me + w_truth * mt))
                    q_curr = (1.0 - lam_damp) * q_curr + lam_damp * ((1.0 - s_swing) * q_zealot + s_swing * (w_work * qw + w_est * qe + w_truth * qt))

        R_batch = 0.5 * (1.0 + m_curr / np.maximum(q_curr, 1e-5))
        win_batch = (R_batch >= 0.50).astype(np.float64)
        ev_samples.append(win_batch @ ev_vec)

        sum_X0_m += (X0 * m_curr).sum(axis=0)
        M1 += m_curr.sum(axis=0)
        M2 += (m_curr ** 2).sum(axis=0)
        M3 += (m_curr ** 3).sum(axis=0)
        M4 += (m_curr ** 4).sum(axis=0)

        sum_q += q_curr.sum(axis=0)
        sum_qw += qw.sum(axis=0)
        sum_qe += qe.sum(axis=0)
        sum_qt += qt.sum(axis=0)
        sum_Dw += D_w.sum(axis=0)
        sum_De += D_e.sum(axis=0)
        sum_Dt += D_t.sum(axis=0)
        sum_soc_m += soc_m.sum(axis=0)

        sum_R += R_batch.sum(axis=0)
        sum_R_sq += (R_batch ** 2).sum(axis=0)
        sum_win += win_batch.sum(axis=0)

        batch_means_m[b_idx] = m_curr.mean(axis=0)
        batch_vars_m[b_idx] = m_curr.var(axis=0, ddof=1)

    mu_m = M1 / total_N
    m2_raw, m3_raw, m4_raw = M2 / total_N, M3 / total_N, M4 / total_N
    var_m = np.maximum(m2_raw - mu_m ** 2, 1e-12)
    sigma_m = np.sqrt(var_m)
    mu3_c = m3_raw - 3.0 * mu_m * m2_raw + 2.0 * (mu_m ** 3)
    mu4_c = m4_raw - 4.0 * mu_m * m3_raw + 6.0 * (mu_m ** 2) * m2_raw - 3.0 * (mu_m ** 4)

    skew_m = mu3_c / (sigma_m ** 3)
    kurt_m = (mu4_c / (sigma_m ** 4)) - 3.0
    snr_lam = (mu_m ** 2) / var_m
    delta_ng = 4.0 * (mu_m / sigma_m) * skew_m + kurt_m
    U4_exact = 1.0 - m4_raw / (3.0 * (m2_raw ** 2))

    mean_X0 = sum_X0 / total_N
    var_X0 = (sum_X0_sq / total_N) - mean_X0 ** 2
    cov_X0_m = (sum_X0_m / total_N) - mean_X0 * mu_m
    sobol_S1 = (cov_X0_m ** 2) / np.maximum(var_X0 * var_m, 1e-15)

    W_gr = batch_vars_m.mean(axis=0)
    B_gr = batch_size * batch_means_m.var(axis=0, ddof=1)
    R_hat = np.sqrt((((batch_size - 1.0) / batch_size) * W_gr + (1.0 / batch_size) * B_gr) / np.maximum(W_gr, 1e-15))

    mean_R = sum_R / total_N
    var_R = np.maximum((sum_R_sq / total_N) - mean_R ** 2, 1e-10)
    sigma_R = np.sqrt(var_R)
    win_prob = sum_win / total_N

    R_emp = df_states["R_emp_pct"].values / 100.0
    sigma_emp = 0.025
    brier = (win_prob - (R_emp >= 0.50).astype(float)) ** 2
    z_val = (R_emp - mean_R) / sigma_R
    phi_z = np.exp(-0.5 * z_val ** 2) / np.sqrt(2.0 * np.pi)
    Phi_z = 0.5 * (1.0 + erf(z_val / np.sqrt(2.0)))
    crps = sigma_R * (z_val * (2.0 * Phi_z - 1.0) + 2.0 * phi_z - 1.0 / np.sqrt(np.pi))
    kl_div = np.log(sigma_emp / sigma_R) + (var_R + (mean_R - R_emp) ** 2) / (2.0 * sigma_emp ** 2) - 0.5

    st_df = df_states[["State", "Tier", "EV", "R_emp_pct", "b_k", "s_swing", "E_petro", "Spinodal_Vulnerability_Idx"]].copy()
    st_df["mu_m"] = mu_m
    st_df["sigma_m"] = sigma_m
    st_df["soc_m_field"] = sum_soc_m / total_N
    st_df["SNR_lambda"] = snr_lam
    st_df["U4_Binder"] = U4_exact
    st_df["D_work"] = sum_Dw / total_N
    st_df["D_est"] = sum_De / total_N
    st_df["D_truth"] = sum_Dt / total_N
    st_df["Turnout_q_pct"] = (sum_q / total_N) * 100.0
    st_df["Sim_R_pct"] = mean_R * 100.0
    st_df["Sigma_R_pct"] = sigma_R * 100.0
    st_df["Win_Prob_Rep_pct"] = win_prob * 100.0
    st_df["Sobol_S1_X0"] = sobol_S1
    st_df["Gelman_Rubin_Rhat"] = R_hat
    st_df["Brier_Score"] = brier
    st_df["CRPS"] = crps
    st_df["KL_Div"] = kl_div

    st_df.to_csv(f"{OUTPUT_DIR}/03_State_Posterior_{cfg['label']}.csv", index=False)

    all_ev = np.concatenate(ev_samples)
    bg_mask = df_states["Tier"].values == "Battleground"
    lean_mask = df_states["Tier"].values == "Lean"
    blue_mask = df_states["Tier"].values == "Solid Blue"
    nat_R = float((mean_R * ev_vec).sum() / 535.0)
    dem_senate = float(np.clip(35.0 + (1.0 - win_prob[senate_mask]).sum(), 40.0, 62.0))

    summary_row = {
        "Regime_ID": regime_id,
        "Regime_Label": cfg["label"],
        "Short_Title": cfg["short_title"],
        "J_inter": J_inter,
        "Total_Trials": total_N,
        "Batches": n_batches,
        "Mean_Sobol_S1": float(sobol_S1.mean()),
        "Max_Gelman_Rubin_Rhat": float(R_hat.max()),
        "BG_Turnout_q_pct": float((sum_q[bg_mask] / total_N).mean() * 100.0),
        "BG_Rep_Share_R_pct": float(mean_R[bg_mask].mean() * 100.0),
        "Lean_Rep_Share_R_pct": float(mean_R[lean_mask].mean() * 100.0),
        "Blue_Rep_Share_R_pct": float(mean_R[blue_mask].mean() * 100.0),
        "States_Flipped_Count": int(((win_prob >= 0.50) != (R_emp >= 0.50)).sum()),
        "Exp_Dem_Senate": dem_senate,
        "Exp_Rep_Senate": 100.0 - dem_senate,
        "Exp_Rep_EV": float(all_ev.mean()),
        "Std_Rep_EV": float(all_ev.std()),
        "Mean_U4_Binder": float(U4_exact.mean()),
        "Mean_KL_Div": float(kl_div.mean()),
        "Elapsed_Sec": round(time.time() - t0, 2)
    }
    print(f"  [DONE] {cfg['short_title']:<32} | J_inter={J_inter:.2f} | BG_q*={summary_row['BG_Turnout_q_pct']:5.1f}% | "
          f"BG_R={summary_row['BG_Rep_Share_R_pct']:5.1f}% | E[EV]={summary_row['Exp_Rep_EV']:5.1f} | "
          f"S1={summary_row['Mean_Sobol_S1']:.2e}")
    return summary_row, st_df, all_ev

# ==============================================================================
# STEP 3: Execute All 6 Hypothesis 2 Regimes & Export LaTeX Table
# ==============================================================================
df_states, W_inter = build_us50_spatial_network()

H2_REGIMES = [
    {
        "label": "H2_Regime1_2018_Split_Congress_NoSpinodal",
        "short_title": "1. 2018 Split (No Spinodal)",
        "mode": "2018_split", "J_intra": 0.62, "J_inter": 0.12, "lam_damp": 0.68, "q_zealot": 0.92
    },
    {
        "label": "H2_Regime2_2026_Midterm_Bicameral_Flip",
        "short_title": "2. 2026 Bicameral Flip",
        "mode": "2026_flip", "J_intra": 0.62, "J_inter": 0.12, "lam_damp": 0.68, "q_zealot": 0.92
    },
    {
        "label": "H2_Regime3_Post2026_Uncoupled_Jinter0",
        "short_title": "3. Post-2026 Uncoupled (J_int=0)",
        "mode": "post_2026", "J_intra": 0.62, "J_inter": 0.00, "lam_damp": 0.70, "q_zealot": 0.90,
        "alpha_mem": 0.90, "gamma_diss": 0.06, "ep_work_pen": 0.08, "ep_est_pen": 0.72, "petro_weight": 0.90,
        "stages": [(0.35, 0.25, 0.45, 1.40), (0.85, 0.65, 1.25, 1.30), (1.10, 1.15, 1.15, 1.25)]
    },
    {
        "label": "H2_Regime4_Post2026_Coupled_Petro_EuroLeak50",
        "short_title": "4. Coupled Contagion (J_int=0.14)",
        "mode": "post_2026", "J_intra": 0.62, "J_inter": 0.14, "lam_damp": 0.70, "q_zealot": 0.90,
        "alpha_mem": 0.90, "gamma_diss": 0.06, "ep_work_pen": 0.08, "ep_est_pen": 0.72, "petro_weight": 0.90,
        "stages": [(0.35, 0.25, 0.45, 1.40), (0.85, 0.65, 1.25, 1.30), (1.10, 1.15, 1.15, 1.25)]
    },
    {
        "label": "H2_Regime5_Post2026_HyperCoupled_NATO_Petro_Split",
        "short_title": "5. Strong Contagion (J_int=0.22)",
        "mode": "post_2026", "J_intra": 0.68, "J_inter": 0.22, "lam_damp": 0.70, "q_zealot": 0.90,
        "alpha_mem": 0.92, "gamma_diss": 0.06, "ep_work_pen": 0.08, "ep_est_pen": 0.85, "petro_weight": 1.15,
        "stages": [(0.45, 0.35, 0.50, 1.45), (0.95, 0.85, 1.30, 1.35), (1.20, 1.30, 1.25, 1.35)]
    },
    {
        "label": "H2_Regime6_Post2026_BRICS_Sacrifice_ReverseContagion",
        "short_title": "6. BRICS+100% Leak Contagion",
        "mode": "post_2026", "J_intra": 0.65, "J_inter": 0.18, "lam_damp": 0.70, "q_zealot": 0.90,
        "alpha_mem": 0.92, "gamma_diss": 1.45, "ep_work_pen": 1.28, "ep_est_pen": 0.92, "petro_weight": 0.45,
        "stages": [(0.45, 1.40, 1.50, 0.50), (1.00, 1.55, 1.60, 0.85), (1.15, 1.70, 1.70, 1.00)]
    }
]

summary_rows = []
state_dfs = {}
ev_arrays = {}

for idx, cfg in enumerate(H2_REGIMES, start=1):
    s_row, st_df, ev_arr = run_hypothesis2_6batch_mc(idx, cfg, df_states, W_inter)
    summary_rows.append(s_row)
    state_dfs[cfg["label"]] = st_df
    ev_arrays[cfg["label"]] = ev_arr

df_summary = pd.DataFrame(summary_rows)
df_summary.to_csv(f"{OUTPUT_DIR}/04_Hypothesis2_Global_6Regimes_Summary.csv", index=False)

# Export LaTeX Table Snippet
latex_h2 = [
    "% Auto-generated LaTeX Table for Hypothesis 2 (Spinodal Instability & Spatial Coupling Contagion)",
    "\\begin{table}[htbp]",
    "\\centering",
    "\\caption{Hypothesis 2 Verification: Spinodal Contagion and Inter-State Coupling Across 50 U.S. States (6 Batches).}",
    "\\label{tab:hypothesis2_spatial_contagion}",
    "\\scriptsize",
    "\\begin{tabular}{lrrrrrrrrr}",
    "\\toprule",
    "Regime Specification & $J_{\\mathrm{inter}}$ & $\\bar{q}_{\\mathrm{BG}}^*$ (\\%) & $\\bar{R}_{\\mathrm{BG}}$ (\\%) & $\\bar{R}_{\\mathrm{Lean}}$ (\\%) & $\\bar{R}_{\\mathrm{Blue}}$ (\\%) & $\\mathbb{E}[\\mathrm{EV}_{\\mathrm{Rep}}]$ & Flipped States & $\\bar{S}_1(X_0)$ & $\\max \\hat{R}_k$ \\\\",
    "\\midrule"
]
for _, r in df_summary.iterrows():
    latex_h2.append(
        f"{r['Short_Title']} & ${r['J_inter']:.2f}$ & ${r['BG_Turnout_q_pct']:.2f}\\%$ & "
        f"${r['BG_Rep_Share_R_pct']:.2f}\\%$ & ${r['Lean_Rep_Share_R_pct']:.2f}\\%$ & ${r['Blue_Rep_Share_R_pct']:.2f}\\%$ & "
        f"${r['Exp_Rep_EV']:.1f}$ & ${r['States_Flipped_Count']}$ & ${r['Mean_Sobol_S1']:.2e}$ & ${r['Max_Gelman_Rubin_Rhat']:.6f}$ \\\\"
    )
latex_h2 += ["\\bottomrule", "\\end{tabular}", "\\end{table}"]
with open(f"{OUTPUT_DIR}/05_Hypothesis2_LaTeX_Table_Snippet.tex", "w") as f:
    f.write("\n".join(latex_h2))

# ==============================================================================
# STEP 4: Generate Figures & Master Dashboard for Hypothesis 2
# ==============================================================================

# --- FIGURE 1: Spatial Coupling Matrix W_inter & Contagion Amplification ---
fig1, axes1 = plt.subplots(1, 3, figsize=(18.5, 5.4))

im = axes1[0].imshow(W_inter, cmap='magma', aspect='auto')
fig1.colorbar(im, ax=axes1[0], label="Inter-State Coupling Weight W_inter,kl")
axes1[0].set_xlabel("Destination State Index l (1..50)", fontweight='bold')
axes1[0].set_ylabel("Origin State Index k (1..50)", fontweight='bold')
axes1[0].set_title("(A) 50-State Row-Normalized Similarity Matrix W_inter", fontsize=10.5, fontweight='bold')

st_unc = state_dfs["H2_Regime3_Post2026_Uncoupled_Jinter0"]
st_cpl = state_dfs["H2_Regime4_Post2026_Coupled_Petro_EuroLeak50"]
st_hyp = state_dfs["H2_Regime5_Post2026_HyperCoupled_NATO_Petro_Split"]

axes1[1].scatter(st_unc["Spinodal_Vulnerability_Idx"], st_cpl["Sim_R_pct"] - st_unc["Sim_R_pct"],
                 color='#d62728', s=50, alpha=0.85, label="Coupled (J_int=0.14) minus Uncoupled (J_int=0)")
axes1[1].scatter(st_unc["Spinodal_Vulnerability_Idx"], st_hyp["Sim_R_pct"] - st_unc["Sim_R_pct"],
                 color='#9467bd', s=50, alpha=0.85, marker='^', label="Strong Contagion (J_int=0.22) minus Uncoupled")
axes1[1].axhline(0.0, color='black', ls='--')
axes1[1].set_xlabel("Spinodal Vulnerability Index s_swing / (|b_k| + 0.05)", fontweight='bold')
axes1[1].set_ylabel("Spatial Contagion Vote-Share Amplification (%)", fontweight='bold')
axes1[1].set_title("(B) Spinodal Amplification in High-Vulnerability Battlegrounds", fontsize=10.5, fontweight='bold')
axes1[1].legend(fontsize=8)
axes1[1].grid(alpha=0.25)

st_18 = state_dfs["H2_Regime1_2018_Split_Congress_NoSpinodal"]
axes1[2].scatter(st_18["Sim_R_pct"], st_18["Turnout_q_pct"], color='#1f77b4', s=42, alpha=0.8, label="1. 2018 Split Congress (High-Turnout Plate)")
axes1[2].scatter(st_cpl["Sim_R_pct"], st_cpl["Turnout_q_pct"], color='#d62728', s=45, alpha=0.85, label="4. 2026 Coupled Spinodal V-Locus")
axes1[2].axvline(50.0, color='black', ls='--', lw=1.2)
axes1[2].set_xlabel("Simulated Republican Two-Party Share R_k (%)", fontweight='bold')
axes1[2].set_ylabel("Active Voter Turnout q_k* (%)", fontweight='bold')
axes1[2].set_title("(C) 50-State V-Shaped Spinodal Trough (2026) vs. 2018 Plateau", fontsize=10.5, fontweight='bold')
axes1[2].legend(fontsize=8, loc='lower left')
axes1[2].grid(alpha=0.25)

plt.tight_layout()
fig1.savefig(f"{OUTPUT_DIR}/Fig1_Hypothesis2_Spatial_Network_and_VLocus.png")
plt.close(fig1)

# --- FIGURE 2: 50-State Ordered Cascades & Jacobian Amplification ---
fig2, axes2 = plt.subplots(1, 3, figsize=(18.5, 5.4))

order_idx = np.argsort(st_cpl["Sim_R_pct"].values)
axes2[0].plot(range(50), st_18["Sim_R_pct"].values[order_idx], 'o--', color='#1f77b4', ms=4, label="1. 2018 Split Congress")
axes2[0].plot(range(50), state_dfs["H2_Regime2_2026_Midterm_Bicameral_Flip"]["Sim_R_pct"].values[order_idx], 's--', color='#2ca02c', ms=4, label="2. 2026 Midterm Flip")
axes2[0].plot(range(50), st_cpl["Sim_R_pct"].values[order_idx], '^-', color='#d62728', ms=4.5, label="4. Post-2026 Coupled Contagion")
axes2[0].plot(range(50), state_dfs["H2_Regime6_Post2026_BRICS_Sacrifice_ReverseContagion"]["Sim_R_pct"].values[order_idx], 'v-', color='#9467bd', ms=4, label="6. BRICS/100% Leak Reverse Cascade")
axes2[0].axhline(50.0, color='black', ls='--', lw=1.2)
axes2[0].set_xlabel("50 U.S. States (Ranked by Regime 4 Share)", fontweight='bold')
axes2[0].set_ylabel("Effective Republican Two-Party Share R_k (%)", fontweight='bold')
axes2[0].set_title("(A) 50-State Electoral Cascade Profiles", fontsize=10.5, fontweight='bold')
axes2[0].legend(fontsize=7.8, loc='upper left')
axes2[0].grid(alpha=0.25)

# Jacobian inverse-turnout amplification factor 1 / (2 q_k*)
axes2[1].scatter(st_cpl["b_k"], 1.0 / (2.0 * (st_cpl["Turnout_q_pct"] / 100.0)), c=st_cpl["E_petro"], cmap='coolwarm', s=55, edgecolors='k')
axes2[1].axvline(0.0, color='gray', ls=':')
axes2[1].set_xlabel("Historical State Partisan Baseline b_k", fontweight='bold')
axes2[1].set_ylabel("Jacobian Sensitivity dR_k / dm_k = 1 / (2 q_k*)", fontweight='bold')
axes2[1].set_title("(B) Inverse-Turnout Jacobian Amplification Peak at |b_k| ~ 0", fontsize=10.5, fontweight='bold')
axes2[1].grid(alpha=0.25)

x_pos = np.arange(len(df_summary))
axes2[2].plot(x_pos, df_summary["BG_Rep_Share_R_pct"], 'o-', color='#d62728', lw=2.2, ms=7, label="Battleground Tier (7 States)")
axes2[2].plot(x_pos, df_summary["Lean_Rep_Share_R_pct"], 's-', color='#ff7f0e', lw=2.0, ms=6, label="Lean Tier (9 States)")
axes2[2].plot(x_pos, df_summary["Blue_Rep_Share_R_pct"], '^-', color='#1f77b4', lw=2.0, ms=6, label="Solid Blue Tier (14 States)")
axes2[2].axhline(50.0, color='black', ls='--')
axes2[2].set_xticks(x_pos)
axes2[2].set_xticklabels(df_summary["Short_Title"], rotation=25, ha='right', fontsize=8)
axes2[2].set_ylabel("Mean Effective Republican Vote Share (%)", fontweight='bold')
axes2[2].set_title("(C) Tier-Level Contagion Progression Across Regimes", fontsize=10.5, fontweight='bold')
axes2[2].legend(fontsize=8)
axes2[2].grid(alpha=0.25)

plt.tight_layout()
fig2.savefig(f"{OUTPUT_DIR}/Fig2_Hypothesis2_50State_Cascades_and_Jacobian.png")
plt.close(fig2)

# --- FIGURE 3: MASTER DASHBOARD FOR HYPOTHESIS 2 ---
fig3 = plt.figure(figsize=(20, 12))
gs = fig3.add_gridspec(2, 3, hspace=0.35, wspace=0.25)
palette = ['#1f77b4', '#2ca02c', '#ff7f0e', '#d62728', '#8c564b', '#9467bd']

axA = fig3.add_subplot(gs[0, 0])
axA.scatter(st_18["Sim_R_pct"], st_18["Turnout_q_pct"], color='#1f77b4', s=38, alpha=0.8, label="1. 2018 Split Congress")
axA.scatter(st_cpl["Sim_R_pct"], st_cpl["Turnout_q_pct"], color='#d62728', s=42, alpha=0.85, label="4. 2026 Coupled Contagion")
axA.scatter(st_hyp["Sim_R_pct"], st_hyp["Turnout_q_pct"], color='#8c564b', s=38, alpha=0.8, marker='^', label="5. Strong Contagion (J_int=0.22)")
axA.axvline(50.0, color='black', ls='--')
axA.set_xlabel("Effective Republican Share R_k (%)", fontweight='bold')
axA.set_ylabel("Active Voter Turnout q_k* (%)", fontweight='bold')
axA.set_title("(A) Phase-Space V-Locus vs. 2018 High-Turnout Plateau", fontsize=11, fontweight='bold')
axA.legend(fontsize=7.8)
axA.grid(alpha=0.25)

axB = fig3.add_subplot(gs[0, 1])
axB.plot(range(50), st_18["Sim_R_pct"].values[order_idx], 'o--', color='#1f77b4', ms=3.5, label="1. 2018 Split")
axB.plot(range(50), st_unc["Sim_R_pct"].values[order_idx], 's:', color='#ff7f0e', ms=3.5, label="3. 2026 Uncoupled (J_int=0)")
axB.plot(range(50), st_cpl["Sim_R_pct"].values[order_idx], '^-', color='#d62728', ms=4.0, label="4. 2026 Coupled (J_int=0.14)")
axB.plot(range(50), st_hyp["Sim_R_pct"].values[order_idx], 'd-', color='#8c564b', ms=3.8, label="5. Strong Coupled (J_int=0.22)")
axB.axhline(50.0, color='black', ls='--')
axB.set_xlabel("50 U.S. States (Ranked)", fontweight='bold')
axB.set_ylabel("Effective Rep. Share R_k (%)", fontweight='bold')
axB.set_title("(B) Spatial Coupling Amplification Across 50 States", fontsize=11, fontweight='bold')
axB.legend(fontsize=7.8)
axB.grid(alpha=0.25)

axC = fig3.add_subplot(gs[0, 2])
for idx, cfg in enumerate(H2_REGIMES):
    ev_a = ev_arrays[cfg["label"]]
    axC.hist(ev_a, bins=26, alpha=0.45, color=palette[idx], label=f"{cfg['short_title']} ({ev_a.mean():.1f} EV)", density=True)
axC.axvline(270, color='black', ls='--', lw=2.0, label="270 EV Majority")
axC.set_xlabel("Republican Electoral College Votes", fontweight='bold')
axC.set_ylabel("Posterior Density", fontweight='bold')
axC.set_title("(C) Electoral College Posterior Histograms", fontsize=11, fontweight='bold')
axC.legend(fontsize=7.2, loc='upper left')
axC.grid(alpha=0.25)

axD = fig3.add_subplot(gs[1, 0])
axD.scatter(st_cpl["b_k"], st_cpl["U4_Binder"], color='#d62728', s=42, alpha=0.85, label="4. Coupled (J_int=0.14)")
axD.scatter(st_hyp["b_k"], st_hyp["U4_Binder"], color='#8c564b', s=42, alpha=0.85, marker='^', label="5. Strong Coupled (J_int=0.22)")
axD.axhline(0.0, color='black', ls='--', label="Spinodal Boundary (U4 < 0)")
axD.set_xlabel("Partisan Baseline b_k", fontweight='bold')
axD.set_ylabel("Fourth-Order Binder Cumulant U4,k", fontweight='bold')
axD.set_title("(D) Migration of Spinodal Zone (U4 < 0) into Solid Blue Tier", fontsize=11, fontweight='bold')
axD.legend(fontsize=8)
axD.grid(alpha=0.25)

axE = fig3.add_subplot(gs[1, 1])
for idx, cfg in enumerate(H2_REGIMES):
    st = state_dfs[cfg["label"]]
    axE.plot(range(1, 51), st["Sobol_S1_X0"] * 100.0, 'o-', ms=3.2, lw=1.1, color=palette[idx],
             label=f"{cfg['short_title']} (Rhat={st['Gelman_Rubin_Rhat'].max():.5f})")
axE.axhline(0.01, color='red', ls='--', lw=1.2, label="0.01% Bound")
axE.set_xlabel("State Index k (1 to 50)", fontweight='bold')
axE.set_ylabel("First-Order Sobol Index S1(X0) [%]", fontweight='bold')
axE.set_title("(E) Initial-Prior Invariance Proof (6 Batches)", fontsize=11, fontweight='bold')
axE.legend(fontsize=7.0, loc='upper right')
axE.grid(alpha=0.25)

axF = fig3.add_subplot(gs[1, 2])
axF.axis('off')
t_rows = []
for _, r in df_summary.iterrows():
    t_rows.append([
        r["Short_Title"],
        f"{r['J_inter']:.2f}",
        f"{r['BG_Turnout_q_pct']:.1f}%",
        f"{r['BG_Rep_Share_R_pct']:.1f}%",
        f"{r['Exp_Rep_EV']:.1f}",
        f"{r['Mean_Sobol_S1']:.1e}",
        f"{r['Max_Gelman_Rubin_Rhat']:.5f}"
    ])
cols = ["Regime (6 Batches)", "J_int", "BG q*", "BG R%", "E[EV]", "Sobol S1", "Max Rhat"]
tbl = axF.table(cellText=t_rows, colLabels=cols, loc='center', cellLoc='center')
tbl.auto_set_font_size(False)
tbl.set_fontsize(7.8)
tbl.scale(1.06, 1.70)
axF.set_title("(F) Hypothesis 2 Monte Carlo Verification Summary", fontsize=11, fontweight='bold')

plt.suptitle(f"Hypothesis 2 Master Verification Dashboard: Spinodal Contagion & Spatial Coupling ({N_TRIALS_PER_REGIME:,} Trials/Regime)",
             fontsize=12.5, fontweight='bold', y=0.98)
fig3.savefig(f"{OUTPUT_DIR}/Fig3_Hypothesis2_Master_Dashboard.png", bbox_inches='tight')
plt.close(fig3)

# ==============================================================================
# STEP 5: ZIP Compression & Automatic Google Colab Download
# ==============================================================================
zip_h2 = "Hypothesis2_Spinodal_Spatial_Contagion_Verification.zip"
with zipfile.ZipFile(zip_h2, 'w', zipfile.ZIP_DEFLATED) as zf:
    for root, _, files in os.walk(OUTPUT_DIR):
        for f in files:
            zf.write(os.path.join(root, f), arcname=f)

print("=" * 82)
print(f"[SUCCESS] Created Hypothesis 2 archive: {zip_h2}")
print("=" * 82)

try:
    from google.colab import files
    files.download(zip_h2)
except ImportError:
    print(f"[INFO] Saved locally at: ./{zip_h2}")
```

```python
# ==============================================================================
# Google Colab Complete Verification Suite for HYPOTHESIS 3:
# Historical Split-Congress Backtest (2018-2020 Order A vs. Order B) vs.
# 2026-2028 Bicameral Gridlock, Super-State (BRICS/AI) Pacts, Iran/Petro Policy,
# and Multi-Stage Epstein Leak Sequences (6-Batch Vectorized Monte Carlo)
# ==============================================================================
import os
import time
import zipfile
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.special import erf

# Publication-grade plotting configuration
plt.rcParams['figure.dpi'] = 150
plt.rcParams['savefig.dpi'] = 300
plt.rcParams['font.sans-serif'] = 'DejaVu Sans'
plt.rcParams['axes.edgecolor'] = '#222222'
plt.rcParams['axes.linewidth'] = 0.85

OUTPUT_DIR = "Hypothesis3_2018_vs_2026_SuperState_Results"
os.makedirs(OUTPUT_DIR, exist_ok=True)

# ==============================================================================
# MONTE CARLO TRIAL CONFIGURATION (Strictly 6 Independent Batches / Chains)
# - Default: 1,200,000 trials per regime (6 batches x 200,000 trials)
# - For 10,000,000 trials per regime, set N_TRIALS_PER_REGIME = 10_000_000
# ==============================================================================
N_TRIALS_PER_REGIME = 1_200_000  # Set between 1_000_000 and 10_000_000
N_BATCHES = 6                    # 6-way split for memory safety & Gelman-Rubin R_hat
SEED_BASE = 20261103

print("=" * 82)
print(f"[HYPOTHESIS 3 SUITE] Initializing {N_TRIALS_PER_REGIME:,} trials/regime ({N_BATCHES} batches)")
print("=" * 82)

# ==============================================================================
# 1. Historical (2018-2020) vs. Current (2024-2026) Gallup & Institutional Data
# ==============================================================================
def generate_empirical_comparison_tables():
    # Table 1A: 2018-2020 Split Congress (Dem House 235-199, Rep Senate 53-47)
    data_2018_2020 = [
        ("2018-01", 38.0, 57.0, -19.0, 35.0, 20.0, 0.15, "Tax Cuts & Deregulation Baseline"),
        ("2018-06", 42.0, 53.0, -11.0, 39.0, 19.0, 0.18, "Conventional Section 301 Tariffs"),
        ("2018-10", 40.0, 54.0, -14.0, 37.0, 21.0, 0.35, "Pre-Midterm Polling (Gallup 40%-54%)"),
        ("2018-11", 38.0, 60.0, -22.0, 34.0, 20.0, 0.40, "Partial Split: Dem House / Rep Senate"),
        ("2019-01", 37.0, 59.0, -22.0, 33.0, 18.0, 0.30, "35-Day Partial Shutdown Shock"),
        ("2019-04", 46.0, 50.0,  -4.0, 41.0, 20.0, 0.15, "Mueller Report Conclusion Bounce"),
        ("2019-10", 41.0, 57.0, -16.0, 36.0, 19.0, 0.28, "House Impeachment Inquiry Onset"),
        ("2020-02", 49.0, 48.0,  +1.0, 44.0, 23.0, 0.12, "Senate Acquittal & 49% Approval Peak"),
    ]
    df_2018 = pd.DataFrame(data_2018_2020, columns=[
        "Year_Month", "Approve_pct", "Disapprove_pct", "Net_Margin_pct",
        "Ind_Approve_pct", "Cong_Approve_pct", "Protest_Index", "Political_Event"
    ])
    df_2018["Calibrated_Delta_h"] = 0.0042 * (df_2018["Net_Margin_pct"] - (-14.0)) - 0.048 * df_2018["Protest_Index"]

    # Table 1B: 2024-2026 Deep Geopolitical Shift & Bicameral Flip
    data_2024_2026 = [
        ("2024-12", 49.0, 46.0,  +3.0, 44.0, 19.0, 0.05, "Post-Election Transition Honeymoon"),
        ("2025-01", 47.0, 48.0,  -1.0, 46.0, 21.0, 0.10, "Inauguration Baseline Reference"),
        ("2025-03", 44.5, 52.0,  -7.5, 39.0, 21.0, 0.22, "Ukraine Aid Freeze & Tariff Surge"),
        ("2025-06", 40.0, 57.0, -17.0, 33.0, 18.0, 0.55, "Iran Strikes & Petro-Automotive Shift"),
        ("2025-11", 36.0, 60.0, -24.0, 25.0, 14.0, 0.68, "China/Russia De-escalation & AI/BRICS Talks"),
        ("2026-03", 36.5, 60.5, -24.0, 26.0, 14.5, 0.65, "Epstein Initial 3% Domestic Disclosure"),
        ("2026-07", 36.0, 61.0, -25.0, 25.5, 14.0, 0.74, "NATO/European Fracture Over Energy & Ukraine"),
        ("2026-10", 35.5, 64.5, -29.0, 24.0, 13.0, 1.00, "Pre-Midterm 500k Protest & Bicameral Flip"),
    ]
    df_2026 = pd.DataFrame(data_2024_2026, columns=[
        "Year_Month", "Approve_pct", "Disapprove_pct", "Net_Margin_pct",
        "Ind_Approve_pct", "Cong_Approve_pct", "Protest_Index", "Political_Event"
    ])
    df_2026["Calibrated_Delta_h"] = 0.0042 * (df_2026["Net_Margin_pct"] - (-1.0)) - 0.048 * df_2026["Protest_Index"]

    df_2018.to_csv(f"{OUTPUT_DIR}/01A_gallup_2018_2020_split_congress_series.csv", index=False)
    df_2026.to_csv(f"{OUTPUT_DIR}/01B_gallup_2024_2026_bicameral_flip_series.csv", index=False)
    return df_2018, df_2026

# ==============================================================================
# 2. 50-State Structural & Industrial Parameters (Petro/Auto vs. NATO/Asia Trade)
# ==============================================================================
def build_us50_parameters():
    # E_petro in [-1, +1]: + indicates fossil-fuel / ICE-auto industrial reliance;
    # - indicates EV/renewable/NATO/Asian supply-chain reliance.
    states_data = [
        ("PA", "Battleground", 19, 48.6, -0.015, 0.58, 0.41, 0.31, 0.28, +0.45, 0),
        ("MI", "Battleground", 15, 47.6, -0.022, 0.58, 0.42, 0.30, 0.28, +0.65, 1),
        ("WI", "Battleground", 10, 48.5, -0.020, 0.58, 0.42, 0.30, 0.28, +0.35, 0),
        ("GA", "Battleground", 16, 49.3, -0.005, 0.56, 0.40, 0.32, 0.28, +0.20, 1),
        ("NC", "Battleground", 16, 50.6, +0.006, 0.56, 0.41, 0.31, 0.28, +0.25, 1),
        ("AZ", "Battleground", 11, 49.2, -0.008, 0.57, 0.39, 0.31, 0.30, +0.10, 0),
        ("NV", "Battleground",  6, 48.1, -0.014, 0.59, 0.40, 0.30, 0.30, +0.15, 0),
        ("MN", "Lean", 10, 41.0, -0.085, 0.55, 0.36, 0.33, 0.31, +0.10, 1),
        ("NH", "Lean",  4, 42.9, -0.070, 0.56, 0.35, 0.33, 0.32, -0.15, 1),
        ("VA", "Lean", 13, 39.0, -0.115, 0.54, 0.34, 0.36, 0.30, -0.45, 1),
        ("ME", "Lean",  4, 41.3, -0.078, 0.57, 0.36, 0.32, 0.32, -0.10, 1),
        ("NM", "Lean",  5, 40.0, -0.095, 0.55, 0.36, 0.33, 0.31, +0.55, 1),
        ("OH", "Lean", 17, 61.7, +0.115, 0.52, 0.46, 0.28, 0.26, +0.60, 1),
        ("IA", "Lean",  6, 62.5, +0.120, 0.52, 0.46, 0.28, 0.26, +0.50, 1),
        ("FL", "Lean", 30, 59.8, +0.090, 0.51, 0.44, 0.29, 0.27, +0.30, 1),
        ("TX", "Lean", 40, 60.7, +0.098, 0.51, 0.45, 0.28, 0.27, +0.90, 1),
        ("AL", "Solid Red",  9, 81.8, +0.340, 0.40, 0.50, 0.26, 0.24, +0.65, 1),
        ("AK", "Solid Red",  3, 65.7, +0.155, 0.50, 0.46, 0.27, 0.27, +0.95, 1),
        ("AR", "Solid Red",  6, 83.8, +0.370, 0.39, 0.51, 0.25, 0.24, +0.60, 1),
        ("ID", "Solid Red",  4, 84.6, +0.395, 0.38, 0.51, 0.25, 0.24, +0.50, 1),
        ("IN", "Solid Red", 11, 72.0, +0.225, 0.45, 0.48, 0.27, 0.25, +0.65, 0),
        ("KS", "Solid Red",  6, 69.7, +0.200, 0.46, 0.47, 0.28, 0.25, +0.55, 1),
        ("KY", "Solid Red",  8, 81.8, +0.345, 0.40, 0.50, 0.26, 0.24, +0.70, 1),
        ("LA", "Solid Red",  8, 73.8, +0.240, 0.44, 0.48, 0.27, 0.25, +0.90, 1),
        ("MS", "Solid Red",  6, 73.0, +0.232, 0.44, 0.48, 0.27, 0.25, +0.65, 1),
        ("MO", "Solid Red", 10, 70.2, +0.205, 0.46, 0.47, 0.28, 0.25, +0.55, 0),
        ("MT", "Solid Red",  4, 71.7, +0.225, 0.45, 0.48, 0.27, 0.25, +0.70, 1),
        ("NE", "Solid Red",  5, 73.8, +0.245, 0.44, 0.48, 0.27, 0.25, +0.55, 1),
        ("ND", "Solid Red",  3, 84.6, +0.402, 0.38, 0.52, 0.24, 0.24, +0.95, 0),
        ("OK", "Solid Red",  7, 84.6, +0.415, 0.37, 0.52, 0.24, 0.24, +0.90, 1),
        ("SC", "Solid Red",  9, 67.7, +0.180, 0.47, 0.47, 0.28, 0.25, +0.55, 1),
        ("SD", "Solid Red",  3, 81.8, +0.340, 0.40, 0.50, 0.26, 0.24, +0.60, 1),
        ("TN", "Solid Red", 11, 80.2, +0.335, 0.40, 0.50, 0.26, 0.24, +0.65, 1),
        ("UT", "Solid Red",  6, 75.8, +0.280, 0.43, 0.49, 0.26, 0.25, +0.50, 0),
        ("WV", "Solid Red",  4, 84.6, +0.460, 0.36, 0.53, 0.24, 0.23, +0.95, 1),
        ("WY", "Solid Red",  3, 84.6, +0.490, 0.35, 0.54, 0.23, 0.23, +0.95, 1),
        ("CA", "Solid Blue", 54, 16.0, -0.390, 0.42, 0.28, 0.38, 0.34, -0.85, 0),
        ("CO", "Solid Blue", 10, 34.4, -0.175, 0.52, 0.32, 0.35, 0.33, -0.25, 1),
        ("CT", "Solid Blue",  7, 26.0, -0.295, 0.46, 0.30, 0.37, 0.33, -0.65, 0),
        ("DE", "Solid Blue",  3, 28.0, -0.265, 0.47, 0.31, 0.36, 0.33, -0.50, 1),
        ("HI", "Solid Blue",  4, 16.0, -0.382, 0.42, 0.28, 0.38, 0.34, -0.75, 0),
        ("IL", "Solid Blue", 19, 28.0, -0.258, 0.47, 0.31, 0.36, 0.33, -0.45, 1),
        ("MD", "Solid Blue", 10, 16.0, -0.400, 0.41, 0.28, 0.39, 0.33, -0.75, 0),
        ("MA", "Solid Blue", 11, 15.0, -0.435, 0.40, 0.27, 0.39, 0.34, -0.85, 1),
        ("NJ", "Solid Blue", 14, 30.0, -0.245, 0.48, 0.31, 0.36, 0.33, -0.55, 1),
        ("NY", "Solid Blue", 28, 21.7, -0.345, 0.44, 0.29, 0.38, 0.33, -0.80, 0),
        ("OR", "Solid Blue",  8, 30.2, -0.225, 0.49, 0.31, 0.35, 0.34, -0.60, 1),
        ("RI", "Solid Blue",  4, 21.9, -0.335, 0.44, 0.29, 0.38, 0.33, -0.70, 1),
        ("VT", "Solid Blue",  3, 15.0, -0.450, 0.40, 0.27, 0.39, 0.34, -0.85, 0),
        ("WA", "Solid Blue", 12, 26.1, -0.282, 0.46, 0.30, 0.36, 0.34, -0.75, 0),
    ]
    df = pd.DataFrame(states_data, columns=[
        "State", "Tier", "EV", "R_emp_pct", "b_k", "s_swing",
        "w_work", "w_est", "w_truth", "E_petro", "Senate_Contest_2026"
    ])
    df.to_csv(f"{OUTPUT_DIR}/02_us50_structural_parameters.csv", index=False)

    b_vec = df["b_k"].values
    diff_mat = np.abs(b_vec[:, None] - b_vec[None, :])
    W_raw = np.exp(-diff_mat / 0.15)
    np.fill_diagonal(W_raw, 0.0)
    W_inter = W_raw / W_raw.sum(axis=1, keepdims=True)
    return df, W_inter

# ==============================================================================
# 3. Continuous-Time Trajectory Simulator for Hypothesis 3 (2018 vs. 2026 Orders)
# ==============================================================================
def simulate_h3_trajectories():
    T_steps = 90
    t_arr = np.arange(T_steps)
    beta = 2.45

    # Compare 4 canonical trajectories in Pennsylvania (b_PA = -0.015, E_petro = +0.45)
    b_pa = -0.015
    s_sw = 0.58
    ww, we, wt = 0.41, 0.31, 0.28
    E_petro = +0.45

    configs = [
        ("2018_OrderA_ImpeachmentFirst", 0.25, (15, 42, 60), 0.15, 0.05, 0.20, 0.25),
        ("2018_OrderB_ShutdownTariffFirst", 0.25, (55, 15, 42), 0.15, 0.05, 0.20, 0.25),
        ("2026_OrderA_EuroLeak50_First", 0.88, (14, 38, 56), 0.78, 0.08, 0.08, 0.85),
        ("2026_OrderB_IranPetroPanic_First", 0.92, (56, 14, 36), 0.78, 0.08, 0.08, 1.00),
        ("2026_OrderC_BRICS_TrumpSacrifice_First", 0.94, (14, 48, 18), 0.95, 1.25, 1.35, 0.60),
    ]

    rows = []
    for label, r_mem, (t_ep, t_pan, t_tri), ep_est_p, ep_work_p, g_diss, petro_w in configs:
        m_c, q_c = -0.06, 0.88
        M_w, M_e, M_t = 0.0, 0.0, 0.0
        tau_rel = 18.0 if "2018" in label else 48.0

        for t in t_arr:
            I_ep  = (1.0 - np.exp(-max(0, t - t_ep)/2.0)) * np.exp(-max(0, t - t_ep)/tau_rel) if t >= t_ep else 0.0
            I_pan = (1.0 - np.exp(-max(0, t - t_pan)/2.0)) * np.exp(-max(0, t - t_pan)/tau_rel) if t >= t_pan else 0.0
            I_tri = (1.0 - np.exp(-max(0, t - t_tri)/2.0)) * np.exp(-max(0, t - t_tri)/tau_rel) if t >= t_tri else 0.0

            soc_m = 0.74 * m_c
            psi_diss = g_diss * (max(0.0, I_tri - 0.25) ** 2)

            if "2018" in label:
                # 2018-2020 Partial Split: Republican Senate buffer keeps D_est > D_t (-0.4621)
                h_w = b_pa + 0.35 + 0.15 * I_pan + 0.08 * I_tri
                D_w = 0.55 + 0.15 * I_pan - 0.05 * I_ep - 0.20 * M_w
                h_e = b_pa - 0.22 + 0.06 * I_pan
                D_e = 0.42 - 0.18 * I_pan - ep_est_p * I_ep - 0.15 * I_tri - 0.20 * M_e
                h_t = b_pa - 0.18 - 0.08 * I_ep
                D_t_f = 0.38 - 0.35 * I_ep - 0.15 * I_pan - 0.20 * M_t
            else:
                # 2026-2028 Bicameral Gridlock + Super-State / Petro / Leak Cascades
                h_w = (b_pa + 0.46 + 0.22 * I_pan + 0.18 * I_tri + 0.25 * petro_w * E_petro * I_pan
                       - 0.55 * psi_diss - 0.45 * ep_work_p * I_ep)
                D_w = (0.62 + 0.28 * I_pan + 0.15 * petro_w * E_petro * I_pan
                       - 0.85 * psi_diss - ep_work_p * I_ep - 0.45 * M_w)
                h_e = b_pa - 0.36 + 0.10 * I_pan + 0.12 * I_tri
                D_e = 0.48 - 0.58 * I_pan - ep_est_p * I_ep - 0.48 * I_tri - 0.52 * M_e
                h_t = b_pa - 0.24 - 0.06 * I_ep + 0.05 * I_pan
                D_t_f = 0.36 - 1.18 * I_ep - 0.45 * I_pan - 0.35 * I_tri - 0.55 * M_t

            aw, ae, at = beta*(soc_m + h_w), beta*(soc_m + h_e), beta*(soc_m + h_t)
            dw, de, dt = np.exp(-beta*D_w) + 2*np.cosh(aw), np.exp(-beta*D_e) + 2*np.cosh(ae), np.exp(-beta*D_t_f) + 2*np.cosh(at)
            mw, qw = 2*np.sinh(aw)/dw, 2*np.cosh(aw)/dw
            me, qe = 2*np.sinh(ae)/de, 2*np.cosh(ae)/de
            mt, qt = 2*np.sinh(at)/dt, 2*np.cosh(at)/dt

            M_w = max(r_mem * M_w, 0.88 * (1.0 - qw))
            M_e = max(r_mem * M_e, 0.90 * (1.0 - qe))
            M_t = max(r_mem * M_t, 0.92 * (1.0 - qt))

            m_c = 0.35 * m_c + 0.65 * ((1.0 - s_sw)*b_pa + s_sw*(ww*mw + we*me + wt*mt))
            q_c = 0.35 * q_c + 0.65 * ((1.0 - s_sw)*0.91 + s_sw*(ww*qw + we*qe + wt*qt))
            R_eff = 50.0 * (1.0 + m_c / max(q_c, 1e-5))

            rows.append({
                "Scenario": label, "Step_t": t, "R_eff_pct": R_eff, "Turnout_q_pct": q_c * 100.0,
                "D_work": D_w, "D_est": D_e, "D_truth": D_t_f,
                "q_work_pct": qw * 100.0, "q_est_pct": qe * 100.0, "q_truth_pct": qt * 100.0
            })

    df_traj = pd.DataFrame(rows)
    df_traj.to_csv(f"{OUTPUT_DIR}/03_hypothesis3_dynamic_trajectories_2018_vs_2026.csv", index=False)
    return df_traj

# ==============================================================================
# 4. 6-Batch Vectorized Monte Carlo Engine for Hypothesis 3 Regimes
# ==============================================================================
def run_h3_monte_carlo_regime(regime_id, cfg, df_states, W_inter, n_trials=N_TRIALS_PER_REGIME, n_batches=N_BATCHES):
    t0 = time.time()
    K = len(df_states)
    batch_size = n_trials // n_batches
    total_N = batch_size * n_batches

    b_k = df_states["b_k"].values[None, :]
    s_swing = df_states["s_swing"].values[None, :]
    w_work = df_states["w_work"].values[None, :]
    w_est = df_states["w_est"].values[None, :]
    w_truth = df_states["w_truth"].values[None, :]
    E_petro = df_states["E_petro"].values[None, :]
    ev_vec = df_states["EV"].values
    senate_mask = df_states["Senate_Contest_2026"].values == 1

    beta = 2.45
    J_intra = 0.62
    J_inter = cfg["J_inter"]
    lam_damp = cfg["lam_damp"]
    q_zealot = cfg["q_zealot"]

    sum_X0 = np.zeros(K, dtype=np.float64)
    sum_X0_sq = np.zeros(K, dtype=np.float64)
    sum_X0_m = np.zeros(K, dtype=np.float64)
    M1, M2, M3, M4 = [np.zeros(K, dtype=np.float64) for _ in range(4)]

    sum_q, sum_qw, sum_qe, sum_qt = [np.zeros(K, dtype=np.float64) for _ in range(4)]
    sum_Dw, sum_De, sum_Dt = [np.zeros(K, dtype=np.float64) for _ in range(3)]
    sum_R, sum_R_sq, sum_win = [np.zeros(K, dtype=np.float64) for _ in range(3)]

    batch_means_m = np.zeros((n_batches, K), dtype=np.float64)
    batch_vars_m = np.zeros((n_batches, K), dtype=np.float64)
    ev_samples = []

    for b_idx in range(n_batches):
        rng = np.random.default_rng(SEED_BASE + regime_id * 1000 + b_idx)

        # Uniform Initial Prior X_0 ~ U(0, 0.2) across all 50 states
        X0 = rng.uniform(0.0, 0.2, size=(batch_size, K))
        m_curr = X0.copy()
        q_curr = np.full((batch_size, K), 0.85, dtype=np.float64)

        sum_X0 += X0.sum(axis=0)
        sum_X0_sq += (X0 ** 2).sum(axis=0)

        if cfg["era"] == "2026_flip":
            xi_prot = rng.gamma(4.5, 0.22, size=(batch_size, 1))
            xi_econ = rng.normal(-0.12, 0.045, size=(batch_size, 1))
            eta_h = rng.normal(0.0, 0.028, size=(batch_size, K))
            eta_D = rng.normal(0.0, 0.030, size=(batch_size, K))
            delta_h_anti = -0.1656

            for _ in range(12):
                soc_m = J_intra * m_curr + J_inter * (m_curr @ W_inter.T)
                h_w = b_k + 0.22 + 0.45 * xi_econ - 0.05 * xi_prot + eta_h
                D_w = 0.15 - 0.38 * xi_prot + eta_D
                h_e = b_k - 0.52 + 0.55 * delta_h_anti - 0.22 * xi_prot + eta_h
                D_e = 0.78 + 0.35 * xi_prot + eta_D
                h_t = b_k - 0.38 + 0.85 * delta_h_anti - 0.28 * xi_prot + eta_h
                D_t = 0.52 + 0.20 * xi_prot + eta_D

                aw, ae, at = beta*(soc_m + h_w), beta*(soc_m + h_e), beta*(soc_m + h_t)
                dw, de, dt = np.exp(-beta*D_w) + 2*np.cosh(aw), np.exp(-beta*D_e) + 2*np.cosh(ae), np.exp(-beta*D_t) + 2*np.cosh(at)
                mw, qw = 2*np.sinh(aw)/dw, 2*np.cosh(aw)/dw
                me, qe = 2*np.sinh(ae)/de, 2*np.cosh(ae)/de
                mt, qt = 2*np.sinh(at)/dt, 2*np.cosh(at)/dt

                m_curr = (1.0 - lam_damp)*m_curr + lam_damp*((1.0 - s_swing)*b_k + s_swing*(w_work*mw + w_est*me + w_truth*mt))
                q_curr = (1.0 - lam_damp)*q_curr + lam_damp*((1.0 - s_swing)*q_zealot + s_swing*(w_work*qw + w_est*qe + w_truth*qt))

        else:
            # 3-Stage Asynchronous Cascade (Handles both 2018-2020 Split and 2026-2028 Bicameral Gridlock)
            xi_war   = rng.gamma(2.5, 0.22, size=(batch_size, 1))
            xi_ep    = rng.gamma(3.0, 0.28, size=(batch_size, 1))
            xi_tri   = rng.gamma(2.8, 0.24, size=(batch_size, 1))
            xi_panic = rng.gamma(3.5, 0.25, size=(batch_size, 1))

            eta_h = rng.normal(0.0, 0.026, size=(batch_size, K))
            eta_D = rng.normal(0.0, 0.028, size=(batch_size, K))

            M_w = np.zeros((batch_size, K), dtype=np.float64)
            M_e = np.zeros((batch_size, K), dtype=np.float64)
            M_t = np.zeros((batch_size, K), dtype=np.float64)

            alpha_mem = cfg["alpha_mem"]
            g_diss = cfg["gamma_diss"]
            ep_w_pen = cfg["ep_work_pen"]
            ep_e_pen = cfg["ep_est_pen"]
            petro_w = cfg["petro_weight"]
            gridlock_scale = cfg["gridlock_scale"] # 0.35 in 2018 Split Congress vs. 1.00 in 2026 Bicameral Flip

            for (ww, we, wt, wp) in cfg["stages"]:
                psi_diss = g_diss * (np.maximum(0.0, wt * xi_tri - 0.40) ** 2)
                for _ in range(4):
                    soc_m = J_intra * m_curr + J_inter * (m_curr @ W_inter.T)
                    delta_h_petro = petro_w * (0.28 * E_petro * wp * xi_panic - 0.20 * (1.0 - E_petro) * ww * xi_war)

                    h_w = (b_k + cfg["h_w0"] + 0.14*ww*xi_war + 0.20*wt*xi_tri + 0.18*wp*xi_panic
                           + delta_h_petro - 0.55*psi_diss - 0.42*ep_w_pen*we*xi_ep + eta_h)
                    D_w = (cfg["D_w0"] + 0.25*wp*xi_panic + 0.15*petro_w*E_petro*wp*xi_panic
                           - 0.85*psi_diss - ep_w_pen*we*xi_ep - 0.38*M_w + eta_D)

                    h_e = b_k + cfg["h_e0"] + 0.08*ww*xi_war + 0.10*wt*xi_tri + 0.08*wp*xi_panic + 0.4*delta_h_petro + eta_h
                    D_e = (cfg["D_e0"] - gridlock_scale*(0.40*ww*xi_war + ep_e_pen*we*xi_ep + 0.46*wt*xi_tri + 0.50*wp*xi_panic)
                           - 0.38*M_e + eta_D)

                    h_t = b_k + cfg["h_t0"] + 0.05*ww*xi_war - 0.06*we*xi_ep + eta_h
                    D_t = (cfg["D_t0"] - gridlock_scale*(0.28*ww*xi_war + 1.15*we*xi_ep + 0.25*wt*xi_tri + 0.38*wp*xi_panic)
                           - 0.42*M_t + eta_D)

                    aw, ae, at = beta*(soc_m + h_w), beta*(soc_m + h_e), beta*(soc_m + h_t)
                    dw, de, dt = np.exp(-beta*D_w) + 2*np.cosh(aw), np.exp(-beta*D_e) + 2*np.cosh(ae), np.exp(-beta*D_t) + 2*np.cosh(at)
                    mw, qw = 2*np.sinh(aw)/dw, 2*np.cosh(aw)/dw
                    me, qe = 2*np.sinh(ae)/de, 2*np.cosh(ae)/de
                    mt, qt = 2*np.sinh(at)/dt, 2*np.cosh(at)/dt

                    M_w = alpha_mem * M_w + (1.0 - alpha_mem) * (1.0 - qw)
                    M_e = alpha_mem * M_e + (1.0 - alpha_mem) * (1.0 - qe)
                    M_t = alpha_mem * M_t + (1.0 - alpha_mem) * (1.0 - qt)

                    m_curr = (1.0 - lam_damp)*m_curr + lam_damp*((1.0 - s_swing)*b_k + s_swing*(w_work*mw + w_est*me + w_truth*mt))
                    q_curr = (1.0 - lam_damp)*q_curr + lam_damp*((1.0 - s_swing)*q_zealot + s_swing*(w_work*qw + w_est*qe + w_truth*qt))

        R_batch = 0.5 * (1.0 + m_curr / np.maximum(q_curr, 1e-5))
        win_batch = (R_batch >= 0.50).astype(np.float64)
        ev_samples.append(win_batch @ ev_vec)

        sum_X0_m += (X0 * m_curr).sum(axis=0)
        M1 += m_curr.sum(axis=0)
        M2 += (m_curr ** 2).sum(axis=0)
        M3 += (m_curr ** 3).sum(axis=0)
        M4 += (m_curr ** 4).sum(axis=0)

        sum_q += q_curr.sum(axis=0)
        sum_qw += qw.sum(axis=0)
        sum_qe += qe.sum(axis=0)
        sum_qt += qt.sum(axis=0)
        sum_Dw += D_w.sum(axis=0)
        sum_De += D_e.sum(axis=0)
        sum_Dt += D_t.sum(axis=0)

        sum_R += R_batch.sum(axis=0)
        sum_R_sq += (R_batch ** 2).sum(axis=0)
        sum_win += win_batch.sum(axis=0)

        batch_means_m[b_idx] = m_curr.mean(axis=0)
        batch_vars_m[b_idx] = m_curr.var(axis=0, ddof=1)

    # Moment & Binder Cumulant Calculations
    mu_m = M1 / total_N
    m2_raw, m3_raw, m4_raw = M2 / total_N, M3 / total_N, M4 / total_N
    var_m = np.maximum(m2_raw - mu_m ** 2, 1e-12)
    sigma_m = np.sqrt(var_m)
    mu3_c = m3_raw - 3.0 * mu_m * m2_raw + 2.0 * (mu_m ** 3)
    mu4_c = m4_raw - 4.0 * mu_m * m3_raw + 6.0 * (mu_m ** 2) * m2_raw - 3.0 * (mu_m ** 4)

    skew_m = mu3_c / (sigma_m ** 3)
    kurt_m = (mu4_c / (sigma_m ** 4)) - 3.0
    snr_lam = (mu_m ** 2) / var_m
    delta_ng = 4.0 * (mu_m / sigma_m) * skew_m + kurt_m
    U4_exact = 1.0 - m4_raw / (3.0 * (m2_raw ** 2))

    # Sobol S1 & 6-Chain Gelman-Rubin R_hat
    mean_X0 = sum_X0 / total_N
    var_X0 = (sum_X0_sq / total_N) - mean_X0 ** 2
    cov_X0_m = (sum_X0_m / total_N) - mean_X0 * mu_m
    sobol_S1 = (cov_X0_m ** 2) / np.maximum(var_X0 * var_m, 1e-15)

    W_gr = batch_vars_m.mean(axis=0)
    B_gr = batch_size * batch_means_m.var(axis=0, ddof=1)
    R_hat = np.sqrt((((batch_size - 1.0) / batch_size) * W_gr + (1.0 / batch_size) * B_gr) / np.maximum(W_gr, 1e-15))

    mean_R = sum_R / total_N
    var_R = np.maximum((sum_R_sq / total_N) - mean_R ** 2, 1e-10)
    sigma_R = np.sqrt(var_R)
    win_prob = sum_win / total_N

    R_emp = df_states["R_emp_pct"].values / 100.0
    sigma_emp = 0.025
    brier = (win_prob - (R_emp >= 0.50).astype(float)) ** 2
    z_val = (R_emp - mean_R) / sigma_R
    phi_z = np.exp(-0.5 * z_val ** 2) / np.sqrt(2.0 * np.pi)
    Phi_z = 0.5 * (1.0 + erf(z_val / np.sqrt(2.0)))
    crps = sigma_R * (z_val * (2.0 * Phi_z - 1.0) + 2.0 * phi_z - 1.0 / np.sqrt(np.pi))
    kl_div = np.log(sigma_emp / sigma_R) + (var_R + (mean_R - R_emp) ** 2) / (2.0 * sigma_emp ** 2) - 0.5

    st_df = df_states[["State", "Tier", "EV", "R_emp_pct", "b_k", "E_petro"]].copy()
    st_df["mu_m"] = mu_m
    st_df["sigma_m"] = sigma_m
    st_df["SNR_lambda"] = snr_lam
    st_df["Skewness"] = skew_m
    st_df["Kurtosis"] = kurt_m
    st_df["Delta_NG"] = delta_ng
    st_df["U4_Binder"] = U4_exact
    st_df["D_work"] = sum_Dw / total_N
    st_df["D_est"] = sum_De / total_N
    st_df["D_truth"] = sum_Dt / total_N
    st_df["q_work_pct"] = (sum_qw / total_N) * 100.0
    st_df["q_est_pct"] = (sum_qe / total_N) * 100.0
    st_df["q_truth_pct"] = (sum_qt / total_N) * 100.0
    st_df["Turnout_q_pct"] = (sum_q / total_N) * 100.0
    st_df["Sim_R_pct"] = mean_R * 100.0
    st_df["Sigma_R_pct"] = sigma_R * 100.0
    st_df["Win_Prob_Rep_pct"] = win_prob * 100.0
    st_df["Sobol_S1_X0"] = sobol_S1
    st_df["Gelman_Rubin_Rhat"] = R_hat
    st_df["Brier_Score"] = brier
    st_df["CRPS"] = crps
    st_df["KL_Div"] = kl_div

    st_df.to_csv(f"{OUTPUT_DIR}/04_state_posterior_{cfg['label']}.csv", index=False)

    all_ev = np.concatenate(ev_samples)
    bg_mask = df_states["Tier"].values == "Battleground"
    nat_R = float((mean_R * ev_vec).sum() / 535.0)
    dem_senate = float(np.clip(35.0 + (1.0 - win_prob[senate_mask]).sum(), 40.0, 62.0))
    dem_house = float(np.clip(np.round(212.0 + 380.0 * ((1.0 - nat_R) - 0.50)), 175.0, 265.0))

    summary_row = {
        "Regime_ID": regime_id,
        "Regime_Label": cfg["label"],
        "Short_Title": cfg["short_title"],
        "Total_Trials": total_N,
        "Batches": n_batches,
        "Prior_Mean_X0": float(mean_X0.mean()),
        "Mean_Sobol_S1": float(sobol_S1.mean()),
        "Max_Gelman_Rubin_Rhat": float(R_hat.max()),
        "BG_Mean_D_work": float((sum_Dw[bg_mask] / total_N).mean()),
        "BG_Mean_D_est": float((sum_De[bg_mask] / total_N).mean()),
        "BG_Mean_D_truth": float((sum_Dt[bg_mask] / total_N).mean()),
        "BG_Turnout_q_pct": float((sum_q[bg_mask] / total_N).mean() * 100.0),
        "BG_Rep_Share_R_pct": float(mean_R[bg_mask].mean() * 100.0),
        "Nat_Turnout_q_pct": float((sum_q / total_N).mean() * 100.0),
        "Nat_Rep_Share_R_pct": nat_R * 100.0,
        "Exp_Dem_Senate": dem_senate,
        "Exp_Rep_Senate": 100.0 - dem_senate,
        "Exp_Dem_House": dem_house,
        "Exp_Rep_House": 435.0 - dem_house,
        "Exp_Rep_EV": float(all_ev.mean()),
        "Std_Rep_EV": float(all_ev.std()),
        "Rep_Majority_Prob_pct": float((all_ev >= 270).mean() * 100.0),
        "Mean_U4_Binder": float(U4_exact.mean()),
        "Mean_Brier": float(brier.mean()),
        "Mean_CRPS": float(crps.mean()),
        "Mean_KL_Div": float(kl_div.mean()),
        "Elapsed_Sec": round(time.time() - t0, 2)
    }
    print(f"  [DONE] {cfg['short_title']:<32} | BG_q*={summary_row['BG_Turnout_q_pct']:5.2f}% | "
          f"BG_R={summary_row['BG_Rep_Share_R_pct']:5.2f}% | E[EV]={summary_row['Exp_Rep_EV']:5.1f} | "
          f"S1={summary_row['Mean_Sobol_S1']:.2e} | Rhat={summary_row['Max_Gelman_Rubin_Rhat']:.5f}")
    return summary_row, st_df, all_ev

# ==============================================================================
# 5. Execute All 6 Regimes Testing Hypothesis 3
# ==============================================================================
df_2018, df_2026 = generate_empirical_comparison_tables()
df_states, W_inter = build_us50_parameters()
df_traj = simulate_h3_trajectories()

H3_REGIMES = [
    {
        "label": "H3_Regime1_2018_Split_OrderA_ImpeachmentFirst",
        "short_title": "1. 2018 Split: Order A (Impeach)",
        "era": "2018_split", "J_inter": 0.12, "lam_damp": 0.68, "q_zealot": 0.92,
        "alpha_mem": 0.25, "gamma_diss": 0.02, "ep_work_pen": 0.04, "ep_est_pen": 0.22,
        "petro_weight": 0.20, "gridlock_scale": 0.32,
        "h_w0": +0.18, "D_w0": +0.55, "h_e0": -0.22, "D_e0": +0.52, "h_t0": -0.18, "D_t0": +0.45,
        "stages": [(0.20, 1.35, 0.15, 0.35), (0.85, 1.05, 0.35, 0.65), (0.80, 0.90, 0.60, 0.75)]
    },
    {
        "label": "H3_Regime2_2018_Split_OrderB_TariffShutdownFirst",
        "short_title": "2. 2018 Split: Order B (Tariff)",
        "era": "2018_split", "J_inter": 0.12, "lam_damp": 0.68, "q_zealot": 0.92,
        "alpha_mem": 0.25, "gamma_diss": 0.02, "ep_work_pen": 0.04, "ep_est_pen": 0.22,
        "petro_weight": 0.20, "gridlock_scale": 0.32,
        "h_w0": +0.18, "D_w0": +0.55, "h_e0": -0.22, "D_e0": +0.52, "h_t0": -0.18, "D_t0": +0.45,
        "stages": [(0.35, 0.20, 0.45, 1.35), (0.85, 0.55, 0.55, 0.90), (0.80, 0.90, 0.60, 0.75)]
    },
    {
        "label": "H3_Regime3_2026_Midterm_Bicameral_Flip",
        "short_title": "3. 2026 Midterm Flip Baseline",
        "era": "2026_flip", "J_inter": 0.12, "lam_damp": 0.68, "q_zealot": 0.92
    },
    {
        "label": "H3_Regime4_2026_OrderA_EuroNATO_Leak50_First",
        "short_title": "4. 2026 Order A (Euro Leak 1st)",
        "era": "2026_post", "J_inter": 0.14, "lam_damp": 0.70, "q_zealot": 0.90,
        "alpha_mem": 0.78, "gamma_diss": 0.08, "ep_work_pen": 0.08, "ep_est_pen": 0.62,
        "petro_weight": 0.85, "gridlock_scale": 1.00,
        "h_w0": +0.50, "D_w0": +0.65, "h_e0": -0.35, "D_e0": +0.44, "h_t0": -0.18, "D_t0": +0.32,
        "stages": [(0.15, 1.35, 0.10, 0.35), (1.25, 1.05, 0.35, 0.75), (1.00, 1.00, 1.25, 1.05)]
    },
    {
        "label": "H3_Regime5_2026_OrderB_IranPetroPanic_First",
        "short_title": "5. 2026 Order B (Petro/Iran 1st)",
        "era": "2026_post", "J_inter": 0.14, "lam_damp": 0.70, "q_zealot": 0.90,
        "alpha_mem": 0.90, "gamma_diss": 0.08, "ep_work_pen": 0.08, "ep_est_pen": 0.72,
        "petro_weight": 1.00, "gridlock_scale": 1.00,
        "h_w0": +0.50, "D_w0": +0.65, "h_e0": -0.35, "D_e0": +0.44, "h_t0": -0.18, "D_t0": +0.32,
        "stages": [(0.35, 0.20, 0.45, 1.45), (0.85, 0.55, 1.35, 1.35), (1.10, 1.20, 1.15, 1.30)]
    },
    {
        "label": "H3_Regime6_2026_OrderC_BRICS_TrumpSacrifice_First",
        "short_title": "6. 2026 Order C (BRICS+100% Leak)",
        "era": "2026_post", "J_inter": 0.14, "lam_damp": 0.70, "q_zealot": 0.90,
        "alpha_mem": 0.92, "gamma_diss": 1.45, "ep_work_pen": 1.25, "ep_est_pen": 0.90,
        "petro_weight": 0.45, "gridlock_scale": 1.00,
        "h_w0": +0.50, "D_w0": +0.65, "h_e0": -0.35, "D_e0": +0.44, "h_t0": -0.18, "D_t0": +0.32,
        "stages": [(0.45, 1.45, 1.55, 0.40), (1.05, 1.55, 1.65, 0.85), (1.15, 1.65, 1.70, 1.05)]
    }
]

summary_list = []
state_results = {}
ev_arrays = {}

for idx, cfg in enumerate(H3_REGIMES, start=1):
    s_row, st_df, ev_arr = run_h3_monte_carlo_regime(idx, cfg, df_states, W_inter)
    summary_list.append(s_row)
    state_results[cfg["label"]] = st_df
    ev_arrays[cfg["label"]] = ev_arr

df_summary = pd.DataFrame(summary_list)
df_summary.to_csv(f"{OUTPUT_DIR}/05_Hypothesis3_Global_6Regimes_Summary.csv", index=False)

# ==============================================================================
# 6. Generate Individual Figures & Master Summary Dashboard
# ==============================================================================

# --- FIGURE 1: Hypothesis 3 Dynamic Trajectories (2018 Split Commutativity vs 2026 Order Bifurcation) ---
fig1, axes1 = plt.subplots(1, 3, figsize=(18.5, 5.4))

traj_colors = {
    "2018_OrderA_ImpeachmentFirst": ("#1f77b4", "--", "2018 Split: Order A (Impeach 1st)"),
    "2018_OrderB_ShutdownTariffFirst": ("#17becf", ":", "2018 Split: Order B (Shutdown 1st)"),
    "2026_OrderA_EuroLeak50_First": ("#ff7f0e", "-.", "2026 Flip: Order A (Euro 50% Leak 1st)"),
    "2026_OrderB_IranPetroPanic_First": ("#d62728", "-", "2026 Flip: Order B (Iran/Petro Panic 1st)"),
    "2026_OrderC_BRICS_TrumpSacrifice_First": ("#9467bd", "-", "2026 Flip: Order C (BRICS+100% Sacrifice 1st)"),
}

for scen, grp in df_traj.groupby("Scenario"):
    col, ls, lbl = traj_colors[scen]
    axes1[0].plot(grp["Step_t"], grp["R_eff_pct"], color=col, ls=ls, lw=2.2, label=lbl)
    axes1[1].plot(grp["Step_t"], grp["D_est"], color=col, ls=ls, lw=2.0, label=lbl)
    axes1[2].plot(grp["Step_t"], grp["q_work_pct"], color=col, ls=ls, lw=2.0, label=lbl)

axes1[0].axhline(50.0, color='black', ls='--', lw=1.2, label="50% Majority Line")
axes1[0].set_xlabel("Relaxation Step t", fontweight='bold')
axes1[0].set_ylabel("Effective Rep. Vote Share R_eff(t) [%]", fontweight='bold')
axes1[0].set_title("(A) Hypothesis 3: Order Commutativity (2018) vs. Bifurcation (2026)", fontsize=10.5, fontweight='bold')
axes1[0].legend(fontsize=7.5, loc='best')
axes1[0].grid(alpha=0.25)

axes1[1].axhline(-0.4621, color='purple', ls='--', lw=1.8, label="Tricritical Spinodal D_t/J = -0.4621")
axes1[1].axhline(0.0, color='gray', ls=':', lw=1.0)
axes1[1].set_xlabel("Relaxation Step t", fontweight='bold')
axes1[1].set_ylabel("Establishment Crystal Field D_est(t)", fontweight='bold')
axes1[1].set_title("(B) Establishment Crystal Field Across Spinodal Boundary", fontsize=10.5, fontweight='bold')
axes1[1].legend(fontsize=7.5, loc='lower left')
axes1[1].grid(alpha=0.25)

axes1[2].set_xlabel("Relaxation Step t", fontweight='bold')
axes1[2].set_ylabel("Working-Class / MAGA Active Turnout q_work*(t) [%]", fontweight='bold')
axes1[2].set_title("(C) MAGA Turnout Lock-In (Order B) vs. BRICS/100% Leak Collapse (Order C)", fontsize=10.5, fontweight='bold')
axes1[2].legend(fontsize=7.5, loc='lower left')
axes1[2].grid(alpha=0.25)

plt.tight_layout()
fig1.savefig(f"{OUTPUT_DIR}/Fig1_Hypothesis3_2018_vs_2026_Order_Trajectories.png")
plt.close(fig1)

# --- FIGURE 2: 50-State Industrial Petro Exposure vs. Order Sensitivity ---
fig2, axes2 = plt.subplots(1, 3, figsize=(18.5, 5.4))

st_18A = state_results["H3_Regime1_2018_Split_OrderA_ImpeachmentFirst"]
st_18B = state_results["H3_Regime2_2018_Split_OrderB_TariffShutdownFirst"]
st_26A = state_results["H3_Regime4_2026_OrderA_EuroNATO_Leak50_First"]
st_26B = state_results["H3_Regime5_2026_OrderB_IranPetroPanic_First"]
st_26C = state_results["H3_Regime6_2026_OrderC_BRICS_TrumpSacrifice_First"]

axes2[0].bar(np.arange(50) - 0.2, st_18B["Sim_R_pct"] - st_18A["Sim_R_pct"], width=0.38, color='#1f77b4',
             label="2018 Split: Order B - Order A Gap (Commutative ~ 0%)")
axes2[0].bar(np.arange(50) + 0.2, st_26B["Sim_R_pct"] - st_26A["Sim_R_pct"], width=0.38, color='#d62728',
             label="2026 Gridlock: Order B - Order A Gap (Non-Commutative > 0%)")
axes2[0].axhline(0.0, color='black', lw=1.0)
axes2[0].set_xlabel("50 U.S. States (Ordered by Tier)", fontweight='bold')
axes2[0].set_ylabel("Order-Induced Vote Share Gap R(Order B) - R(Order A) [%]", fontweight='bold')
axes2[0].set_title("(A) 50-State Proof of Order Non-Commutativity: 2018 vs. 2026", fontsize=10.5, fontweight='bold')
axes2[0].legend(fontsize=8)
axes2[0].grid(alpha=0.25)

axes2[1].scatter(st_26B["E_petro"], st_26B["Sim_R_pct"] - st_26C["Sim_R_pct"],
                 c=st_26B["E_petro"], cmap='coolwarm', s=60, edgecolors='k', lw=0.6)
axes2[1].set_xlabel("State Petroleum & ICE-Auto Exposure Index (E_petro)", fontweight='bold')
axes2[1].set_ylabel("Vote Share Reversal: Order B (Petro) minus Order C (BRICS/Sacrifice) [%]", fontweight='bold')
axes2[1].set_title("(B) Sensitivity to Super-State Dissonance & 100% Epstein Leak", fontsize=10.5, fontweight='bold')
axes2[1].grid(alpha=0.25)

x_idx = np.arange(len(df_summary))
w_bar = 0.25
axes2[2].bar(x_idx - w_bar, df_summary["BG_Mean_D_work"], width=w_bar, color='#d62728', label="D_work (MAGA/Working)")
axes2[2].bar(x_idx, df_summary["BG_Mean_D_est"], width=w_bar, color='#1f77b4', label="D_est (Establishment)")
axes2[2].bar(x_idx + w_bar, df_summary["BG_Mean_D_truth"], width=w_bar, color='#2ca02c', label="D_truth (Independent)")
axes2[2].axhline(-0.4621, color='purple', ls='--', lw=1.8, label="Tricritical D_t = -0.4621")
axes2[2].axhline(0.0, color='black', lw=0.8)
axes2[2].set_xticks(x_idx)
axes2[2].set_xticklabels(df_summary["Short_Title"], rotation=25, ha='right', fontsize=8)
axes2[2].set_ylabel("Battleground Mean Crystal Field D", fontweight='bold')
axes2[2].set_title("(C) Stratum Crystal Fields Across All 6 Regimes", fontsize=10.5, fontweight='bold')
axes2[2].legend(fontsize=7.5)
axes2[2].grid(alpha=0.25)

plt.tight_layout()
fig2.savefig(f"{OUTPUT_DIR}/Fig2_Hypothesis3_Order_Gap_and_Petro_Sensitivity.png")
plt.close(fig2)

# --- FIGURE 3: MASTER SUMMARY DASHBOARD (6-Panel Publication Synthesis) ---
fig3 = plt.figure(figsize=(20, 12))
gs = fig3.add_gridspec(2, 3, hspace=0.35, wspace=0.25)

# Panel A: 2018-20 vs 2024-26 Gallup Approval & Protest Calibration
axA = fig3.add_subplot(gs[0, 0])
axA.plot(range(len(df_2018)), df_2018["Net_Margin_pct"], 'o--', color='#1f77b4', lw=2, label="2018-2020 Net Approval (Split Congress)")
axA.plot(range(len(df_2026)), df_2026["Net_Margin_pct"], 's-', color='#d62728', lw=2.2, label="2024-2026 Net Approval (Bicameral Flip)")
axA.axhline(0, color='gray', ls=':')
axA.set_xticks(range(len(df_2026)))
axA.set_xticklabels(df_2026["Year_Month"], rotation=30, fontsize=8)
axA.set_ylabel("Gallup Net Approval Margin (%)", fontweight='bold')
axA.set_title("(A) Empirical Gallup Trajectory: 2018-20 vs. 2024-26", fontsize=11, fontweight='bold')
axA.legend(fontsize=8)
axA.grid(alpha=0.25)

# Panel B: Dynamic Order Hysteresis in PA
axB = fig3.add_subplot(gs[0, 1])
for scen, grp in df_traj.groupby("Scenario"):
    col, ls, lbl = traj_colors[scen]
    axB.plot(grp["Step_t"], grp["R_eff_pct"], color=col, ls=ls, lw=2.0, label=lbl)
axB.axhline(50.0, color='black', ls='--', lw=1.2)
axB.set_xlabel("Step t", fontweight='bold')
axB.set_ylabel("Effective Rep. Share R_eff(t) [%]", fontweight='bold')
axB.set_title("(B) Hypothesis 3 Dynamic Order Trajectories", fontsize=11, fontweight='bold')
axB.legend(fontsize=7.2, loc='best')
axB.grid(alpha=0.25)

# Panel C: Electoral College Posterior Distributions
axC = fig3.add_subplot(gs[0, 2])
palette = ['#1f77b4', '#17becf', '#2ca02c', '#ff7f0e', '#d62728', '#9467bd']
for idx, cfg in enumerate(H3_REGIMES):
    ev_a = ev_arrays[cfg["label"]]
    axC.hist(ev_a, bins=26, alpha=0.45, color=palette[idx], label=f"{cfg['short_title']} ({ev_a.mean():.1f} EV)", density=True)
axC.axvline(270, color='black', ls='--', lw=2.0, label="270 EV Majority")
axC.set_xlabel("Republican Electoral College Votes", fontweight='bold')
axC.set_ylabel("Posterior Density", fontweight='bold')
axC.set_title("(C) 50-State Electoral College Posterior Distributions", fontsize=11, fontweight='bold')
axC.legend(fontsize=7.2, loc='upper left')
axC.grid(alpha=0.25)

# Panel D: 4th-Order Binder Cumulant U4 vs Turnout q*
axD = fig3.add_subplot(gs[1, 0])
for r_key, col, lbl in [
    ("H3_Regime1_2018_Split_OrderA_ImpeachmentFirst", "#1f77b4", "2018 Split (Order A)"),
    ("H3_Regime5_2026_OrderB_IranPetroPanic_First", "#d62728", "2026 Order B (Petro First)"),
    ("H3_Regime6_2026_OrderC_BRICS_TrumpSacrifice_First", "#9467bd", "2026 Order C (BRICS/100% Leak)")
]:
    st = state_results[r_key]
    axD.scatter(st["Turnout_q_pct"], st["U4_Binder"], color=col, s=36, alpha=0.8, label=lbl)
axD.axhline(2.0/3.0, color='gray', ls=':', label="Gaussian Ordered Limit (2/3)")
axD.axhline(0.0, color='black', ls='--', lw=1.0, label="Spinodal Crossover (U4 < 0)")
axD.set_xlabel("Active Voter Turnout q* (%)", fontweight='bold')
axD.set_ylabel("Fourth-Order Binder Cumulant U4", fontweight='bold')
axD.set_title("(D) Binder Cumulant U4 vs. Active Turnout Across 50 States", fontsize=11, fontweight='bold')
axD.legend(fontsize=8, loc='lower left')
axD.grid(alpha=0.25)

# Panel E: Initial-Value Invariance Proof (Sobol S1 & 6-Batch Gelman-Rubin R_hat)
axE = fig3.add_subplot(gs[1, 1])
for idx, cfg in enumerate(H3_REGIMES):
    st = state_results[cfg["label"]]
    axE.plot(range(1, 51), st["Sobol_S1_X0"] * 100.0, 'o-', ms=3.2, lw=1.1, color=palette[idx],
             label=f"{cfg['short_title']} (Rhat={st['Gelman_Rubin_Rhat'].max():.5f})")
axE.axhline(0.01, color='red', ls='--', lw=1.2, label="0.01% Sensitivity Bound")
axE.set_xlabel("State Index k (1 to 50)", fontweight='bold')
axE.set_ylabel("First-Order Sobol Index S1(X0) [%]", fontweight='bold')
axE.set_title("(E) Initial-Prior Invariance Proof (X0 ~ U(0,0.2), 6 Batches)", fontsize=11, fontweight='bold')
axE.legend(fontsize=7.0, loc='upper right')
axE.grid(alpha=0.25)

# Panel F: Master Quantitative Table
axF = fig3.add_subplot(gs[1, 2])
axF.axis('off')
t_rows = []
for _, r in df_summary.iterrows():
    t_rows.append([
        r["Short_Title"],
        f"{r['BG_Turnout_q_pct']:.1f}%",
        f"{r['BG_Rep_Share_R_pct']:.1f}%",
        f"{r['Exp_Rep_EV']:.1f}",
        f"{r['Mean_Sobol_S1']:.1e}",
        f"{r['Max_Gelman_Rubin_Rhat']:.5f}"
    ])
cols = ["Regime (6 Batches)", "BG q*", "BG R%", "E[EV]", "Sobol S1", "Max Rhat"]
tbl = axF.table(cellText=t_rows, colLabels=cols, loc='center', cellLoc='center')
tbl.auto_set_font_size(False)
tbl.set_fontsize(8.0)
tbl.scale(1.05, 1.70)
axF.set_title("(F) Hypothesis 3 Monte Carlo Verification Summary", fontsize=11, fontweight='bold')

plt.suptitle(f"Hypothesis 3 Master Verification Dashboard: 2018-2020 Split Congress vs. 2026-2028 Super-State & Leak Sequences ({N_TRIALS_PER_REGIME:,} Trials/Regime)",
             fontsize=12.5, fontweight='bold', y=0.98)
fig3.savefig(f"{OUTPUT_DIR}/Fig3_Hypothesis3_Master_Dashboard.png", bbox_inches='tight')
plt.close(fig3)

# ==============================================================================
# 7. Compress All CSVs & Figures into ZIP and Trigger Colab Download
# ==============================================================================
zip_name = "Hypothesis3_2018_vs_2026_SuperState_Verification.zip"
with zipfile.ZipFile(zip_name, 'w', zipfile.ZIP_DEFLATED) as zf:
    for root, _, files in os.walk(OUTPUT_DIR):
        for f in files:
            zf.write(os.path.join(root, f), arcname=f)

print("=" * 82)
print(f"[COMPLETED] Created archive: {zip_name} containing all CSVs and Figures.")
print("=" * 82)

try:
    from google.colab import files
    files.download(zip_name)
except ImportError:
    print(f"[INFO] Running outside Google Colab. Archive saved locally at: ./{zip_name}")
```
